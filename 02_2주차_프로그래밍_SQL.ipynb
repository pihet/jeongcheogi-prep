## Day 8 (월) — C 언어 총정리 ①

- [ ] 연산자 우선순위 & 결합 방향 (특히 `++`, `--`, 삼항, 비트연산)
- [ ] **전위/후위 증감 연산자** — 최다 함정
  - `i++` 는 쓰고 나서 증가, `++i` 는 증가하고 나서 씀
  - `a = i++ + ++i;` 같은 문제는 표를 그려라
- [ ] 비트 연산: `&`, `|`, `^`, `~`, `<<`, `>>` — 2진수 변환 연습
- [ ] 형변환 (int ↔ char ↔ double), 정수 나눗셈 버림 주의 (`7/2 == 3`)
- [ ] 문제 풀이 15문항 + 오답 기록

**함정 체크**
```c
int a = 5;
printf("%d", a++ + ++a);   // a++ → 5(a=6), ++a → 7  ⇒ 12
```

- [ ] 포인터 기본: `int *p = &a;` → `*p` 는 a의 값

## Day 10 (수) — Java ①

- [ ] 클래스/객체/생성자, `this`
- [ ] **상속(extends)과 오버라이딩** — 실기 최빈출
  - 부모 타입 참조변수로 자식 객체를 가리킬 때 **어느 메서드가 호출되는가**
  - 필드는 참조변수 타입, 메서드는 실제 객체 타입을 따른다 (매우 중요)
- [ ] 오버로딩 vs 오버라이딩 차이
- [ ] 접근제어자: private < default < protected < public
- [ ] `static` 변수/메서드, `final`
- [ ] 추상 클래스(abstract) vs 인터페이스(interface)
- [ ] 문제 풀이 12문항

**함정 체크**
```java
class A { void f(){ System.out.print("A"); } }
class B extends A { void f(){ System.out.print("B"); } }
A obj = new B();
obj.f();   // "B"  ← 메서드는 실제 객체를 따라감
```

## Day 12 (금) — Python

- [ ] 자료형: list, tuple, dict, set — **각각의 특징과 차이**
  - list `[]` 변경가능 / tuple `()` 변경불가 / dict `{key:value}` / set `{}` 중복없음
- [ ] **슬라이싱 `a[start:end:step]`** — 최빈출
  - `a[::-1]` 역순, `a[1:3]` 은 인덱스 1,2 (end 미포함)
  - 음수 인덱스: `a[-1]` 마지막 원소
- [ ] 문자열 메서드: `split()`, `join()`, `replace()`, `upper()`, `strip()`
- [ ] 리스트 메서드: `append`, `extend`, `insert`, `pop`, `remove`, `sort`, `reverse`
- [ ] 리스트 컴프리헨션 `[x*2 for x in range(5) if x%2==0]`
- [ ] `range(start, end, step)` — end 미포함
- [ ] 함수, 람다, 클래스(`__init__`, `self`)
- [ ] 문제 풀이 12문항

**함정 체크**
```python
a = [1,2,3,4,5]
print(a[1:4])    # [2, 3, 4]
print(a[::2])    # [1, 3, 5]
print(a[::-1])   # [5, 4, 3, 2, 1]
```

## Day 13 (토) — SQL 집중

- [ ] **DDL**: CREATE / ALTER / DROP / TRUNCATE
  - `CREATE TABLE`, 제약조건(PRIMARY KEY, FOREIGN KEY, UNIQUE, NOT NULL, CHECK, DEFAULT)
  - `ALTER TABLE ~ ADD / MODIFY / DROP COLUMN`
  - **DROP 옵션: CASCADE(연쇄삭제) / RESTRICT(참조중이면 취소)**
- [ ] **DML**: SELECT / INSERT / UPDATE / DELETE
  - `INSERT INTO 테이블(컬럼) VALUES(값);`
  - `UPDATE 테이블 SET 컬럼=값 WHERE 조건;`
  - `DELETE FROM 테이블 WHERE 조건;`
- [ ] **DCL**: GRANT / REVOKE / COMMIT / ROLLBACK / SAVEPOINT
  - `GRANT 권한 ON 테이블 TO 사용자 WITH GRANT OPTION;`
- [ ] SELECT 심화
  - `WHERE` 조건: `BETWEEN`, `IN`, `LIKE '%값%'`, `IS NULL`
  - `GROUP BY` + `HAVING` (**HAVING은 그룹에 대한 조건**)
  - `ORDER BY 컬럼 ASC|DESC`
  - 집계함수: `COUNT`, `SUM`, `AVG`, `MAX`, `MIN`
  - `DISTINCT`
- [ ] **JOIN**: INNER / LEFT OUTER / RIGHT OUTER / FULL OUTER / CROSS / SELF
- [ ] 서브쿼리 (단일행/다중행, `IN`, `EXISTS`, `ANY`, `ALL`)
- [ ] 뷰(VIEW), 인덱스(INDEX) 생성 구문
- [ ] 트리거, 프로시저 기본 구조
- [ ] SQL 문제 20문항 풀이

**SELECT 절 실행 순서 (암기)**
```
FROM → WHERE → GROUP BY → HAVING → SELECT → ORDER BY
```
