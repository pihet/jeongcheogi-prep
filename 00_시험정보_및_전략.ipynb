## 2. 실기 출제 구조 (최근 경향)

| 유형 | 대략 문항 수 | 배점 | 난이도 | 비고 |
|---|---|---|---|---|
| **프로그래밍 (C/Java/Python)** | 6~8문항 | 30~40점 | 상 | **코드 추적 = 실기의 핵심** |
| **SQL** | 2~3문항 | 10~15점 | 중 | 문법 암기하면 거의 먹고 들어감 |
| **소프트웨어 공학 단답형** | 5~7문항 | 25~35점 | 중~상 | 약어/용어 암기 싸움 |
| **네트워크·보안·운영체제** | 3~4문항 | 15~20점 | 중 | 프로토콜, 스케줄링, 공격기법 |

### 여기서 나오는 결론
- 프로그래밍 + SQL 만 다 맞아도 **40~55점**. 여기에 단답형 2~4개면 합격선.
- 즉 **코드 추적을 못 하면 절대 못 붙는다.** 1순위는 무조건 프로그래밍.
- 단답형은 "영문 약어 그대로 쓰기"가 많으므로 **철자까지 정확히** 외워야 함.

## 6. 버릴 것 / 지킬 것

**지킬 것 (절대 포기 금지)**
- C 포인터/배열, Java 상속·오버라이딩, Python 슬라이싱·리스트

**버려도 되는 것 (시간 없으면)**
- 지엽적인 UI 설계 세부 지침
- 잘 안 나오는 미들웨어 세부 분류
- 패키징 도구의 상세 옵션
- 너무 마이너한 신기술 용어 (단, 최근 2년 기출에 나온 건 예외)

## 7. 답안 작성 실전 팁

1. **영문 약어는 대문자로 정확히.** (예: `RARP`, `SNMP`, `DHCP`) 철자 틀리면 오답.
2. 한글 용어를 묻는지 영문을 묻는지 **문제 지시문을 반드시 확인.** ("영문 약어로 쓰시오")
3. 코드 출력 문제는 **반드시 손으로 변수 표를 그려서** 추적. 암산 금지.
4. C의 출력은 개행 여부까지 정확히. `printf("%d ", ...)` 는 공백 포함.
5. 모르는 문제도 **빈칸으로 두지 말 것.** 부분점수는 없지만 찍어서 맞을 수는 있다.
6. 시간 배분: 프로그래밍 문제를 먼저 풀지 말고, **단답형부터 빠르게 훑고** 남은 시간을 코드에 몰아라.
7. 답이 2개 이상 요구되면 **개수를 맞춰 쓸 것.** 하나만 쓰면 오답 처리.
