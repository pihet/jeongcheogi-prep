> 규칙
> 1. 틀린 문제 + **찍어서 맞은 문제**는 전부 여기에 기록
> 2. 기록은 **그날 안에**. 미루면 왜 틀렸는지 기억이 안 난다
> 3. 다음 날 아침 15분은 전날 오답 재풀이
