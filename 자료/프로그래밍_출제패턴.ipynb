## 1. C 언어

### 1-1. 증감 연산자 (최다 함정)

```c
int i = 5;
printf("%d", i++);   // 5 출력 후 i=6
printf("%d", ++i);   // i=7 된 후 7 출력
```

```c
int a = 3;
int b = a++ + ++a;   // a++ -> 3 (a=4), ++a -> 5  =>  b = 8, a = 5
```

### 1-2. 정수 나눗셈 / 나머지

```c
7 / 2      // 3   (소수점 버림)
7 % 2      // 1
-7 / 2     // -3  (0 방향 버림)
7 / 2.0    // 3.5 (실수 연산)
```

### 1-3. 배열과 포인터

```c
int arr[5] = {1,2,3,4,5};
int *p = arr;

arr[2]      ==  *(arr+2)   ==  p[2]   ==  *(p+2)     // 3
*p + 1      // 2  (값에 1 더함)
*(p + 1)    // 2  (다음 원소)
*p++        // 현재 값 반환 후 포인터 이동
(*p)++      // 값을 증가
```

2차원 배열
```c
int a[2][3] = {{1,2,3},{4,5,6}};
a[1][2]  ==  *(*(a+1)+2)  ==  6
```

포인터 배열 vs 배열 포인터
```c
int *p[3];     // 포인터 3개를 담은 배열
int (*p)[3];   // 열이 3개인 배열을 가리키는 포인터
```

### 1-4. Call by Value vs Call by Reference

```c
void swap(int a, int b) { int t=a; a=b; b=t; }        // 원본 안 바뀜
void swap(int *a, int *b){ int t=*a; *a=*b; *b=t; }   // 원본 바뀜  ← 이게 정답
```

### 1-5. 문자열

```c
char s[] = "ABC";     // 실제 크기 4 ('\0' 포함)
strlen(s)             // 3  ('\0' 미포함)
sizeof(s)             // 4  ('\0' 포함)
s[0]                  // 'A' (아스키 65)
s[0] + 1              // 66
printf("%c", s[0]+1)  // 'B'
```

아스키 기억: `'A'=65`, `'a'=97`, `'0'=48`

### 1-6. 비트 연산

```c
5 & 3   // 0101 & 0011 = 0001 = 1
5 | 3   // 0111 = 7
5 ^ 3   // 0110 = 6
~5      // -6   (2의 보수)
5 << 2  // 20   (x4)
20 >> 2 // 5    (/4)
```

### 1-7. 재귀

```c
int f(int n){ return (n<=1) ? 1 : n * f(n-1); }   // 팩토리얼
```
→ 호출 스택을 아래로 그리고, 되돌아오면서 값을 곱해 올라온다.

### 1-8. 구조체

```c
struct P { char *name; int age; };
struct P p = {"kim", 20};
struct P *q = &p;
q->age        ==  (*q).age  ==  20
```

### 1-9. static / 전역 변수

```c
void f(){ static int c = 0; c++; printf("%d", c); }
f(); f(); f();   // 1 2 3   ← static은 값이 유지된다
```

## 2. Java

### 2-1. 상속 + 오버라이딩 (최다 출제)

```java
class A { void f(){ System.out.print("A"); } }
class B extends A { void f(){ System.out.print("B"); } }

A obj = new B();
obj.f();     // "B"   ← 메서드는 실제 객체 타입
```

**필드는 다르다**
```java
class A { int x = 1; }
class B extends A { int x = 2; }
A obj = new B();
System.out.print(obj.x);   // 1   ← 필드는 참조변수 타입
```

정리: **메서드는 객체 따라, 변수(필드)는 타입 따라.**

### 2-2. super / this

```java
class A { A(){ System.out.print("A"); } }
class B extends A { B(){ System.out.print("B"); } }
new B();     // "AB"   ← 부모 생성자가 먼저 호출됨
```

### 2-3. 오버로딩 vs 오버라이딩

| | 오버로딩 | 오버라이딩 |
|---|---|---|
| 위치 | 같은 클래스 | 상속 관계 |
| 이름 | 같음 | 같음 |
| 매개변수 | **다름** | 같음 |
| 반환형 | 상관없음 | 같아야 함 |

### 2-4. String 비교

```java
String a = "abc";
String b = "abc";
String c = new String("abc");

a == b        // true   (문자열 상수 풀)
a == c        // false  (다른 객체)
a.equals(c)   // true   ← 값 비교는 항상 equals
```

### 2-5. static

```java
class C {
    static int cnt = 0;
    C(){ cnt++; }
}
new C(); new C(); new C();
System.out.print(C.cnt);   // 3   ← 모든 인스턴스가 공유
```

### 2-6. 예외 처리

```java
try { return 1; }
catch(Exception e){ return 2; }
finally { System.out.print("F"); }   // ← return이 있어도 반드시 실행
```

### 2-7. 배열 / 문자열 메서드

```java
int[] a = new int[5];   // 기본값 0으로 초기화
a.length                // 배열은 length (괄호 없음)
"abc".length()          // 문자열은 length() (괄호 있음)
"abcde".substring(1,3)  // "bc"  (end 미포함)
"abc".charAt(1)         // 'b'
```

### 2-8. 정수 나눗셈 (C와 동일)

```java
System.out.print(7/2);     // 3
System.out.print(7/2.0);   // 3.5
```

## 3. Python

### 3-1. 슬라이싱 (최다 함정)

```python
a = [1,2,3,4,5]
a[1:4]     # [2, 3, 4]        end 미포함
a[:3]      # [1, 2, 3]
a[2:]      # [3, 4, 5]
a[::2]     # [1, 3, 5]        step 2
a[::-1]    # [5, 4, 3, 2, 1]  역순
a[-1]      # 5
a[-3:-1]   # [3, 4]
```

문자열도 동일
```python
s = "python"
s[1:4]     # "yth"
s[::-1]    # "nohtyp"
```

### 3-2. range

```python
range(5)        # 0 1 2 3 4
range(1,5)      # 1 2 3 4      end 미포함
range(1,10,3)   # 1 4 7
range(5,0,-1)   # 5 4 3 2 1
```

### 3-3. 자료형 4종

| 자료형 | 표기 | 순서 | 변경 | 중복 |
|---|---|---|---|---|
| list | `[1,2,3]` | O | O | O |
| tuple | `(1,2,3)` | O | **X** | O |
| dict | `{'k':1}` | O(3.7+) | O | 키 중복 X |
| set | `{1,2,3}` | **X** | O | **X** |

```python
s = set([1,2,2,3,3,3])
print(s)          # {1, 2, 3}
print(len(s))     # 3
```

### 3-4. 리스트 메서드

```python
a = [1,2,3]
a.append(4)      # [1,2,3,4]      원소 하나 추가
a.extend([5,6])  # [1,2,3,4,5,6]  리스트 이어붙임
a.insert(0,0)    # [0,1,2,3,4,5,6]
a.pop()          # 마지막 원소 꺼내고 제거
a.remove(3)      # 값 3을 제거 (인덱스 아님!)
a.sort()         # 원본 정렬 (반환값 None)
sorted(a)        # 새 리스트 반환
a.reverse()      # 원본 역순
```

**함정**: `a.sort()` 는 None을 반환한다. `b = a.sort()` 하면 b는 None.

### 3-5. 얕은 복사 함정

```python
a = [1,2,3]
b = a           # 같은 객체를 가리킴
b[0] = 99
print(a)        # [99, 2, 3]   ← a도 바뀐다!

c = a[:]        # 복사본
c = a.copy()    # 복사본
```

### 3-6. 문자열 메서드

```python
"a,b,c".split(",")      # ['a','b','c']
"-".join(['a','b'])     # "a-b"
"  ab ".strip()         # "ab"
"abc".upper()           # "ABC"
"abcabc".replace("a","X")  # "XbcXbc"
"abc".find("b")         # 1
```

### 3-7. 리스트 컴프리헨션

```python
[x*2 for x in range(5)]              # [0,2,4,6,8]
[x for x in range(10) if x%3==0]     # [0,3,6,9]
```

### 3-8. 클래스

```python
class A:
    def __init__(self, x):
        self.x = x
    def f(self):
        return self.x * 2

a = A(5)
print(a.f())     # 10
```

상속
```python
class B(A):
    def f(self):
        return self.x * 3
```

### 3-9. 나눗셈

```python
7 / 2      # 3.5   ← 파이썬은 항상 실수!  (C, Java와 다름)
7 // 2     # 3     정수 나눗셈
7 % 2      # 1
2 ** 10    # 1024  거듭제곱
```

**주의: 파이썬의 `/` 는 실수를 반환한다. C/Java와 헷갈리지 말 것.**

### 3-10. print 옵션

```python
print(1, 2)             # "1 2"   (기본 구분자 = 공백)
print(1, 2, sep="-")    # "1-2"
print(1, end="")        # 개행 없음
```

## 4. 언어별 헷갈리는 지점 비교

| 항목 | C | Java | Python |
|---|---|---|---|
| `7/2` | 3 | 3 | **3.5** |
| 정수 나눗셈 | `/` | `/` | `//` |
| 배열 길이 | `sizeof` 계산 | `arr.length` | `len(arr)` |
| 문자열 길이 | `strlen(s)` | `s.length()` | `len(s)` |
| 인덱스 시작 | 0 | 0 | 0 (음수 인덱스 지원) |
| 문자열 비교 | `strcmp` | `equals()` | `==` 로 값 비교 가능 |

## 5. 자주 나오는 알고리즘 코드

- **최댓값/최솟값 찾기** (배열 순회)
- **정렬**: 버블 정렬, 선택 정렬, 삽입 정렬 (교환 횟수 / 한 회전 후 상태 묻는 문제)

- **소수 판별 / 약수 구하기**
- **팩토리얼 / 피보나치 (재귀)**
- **문자열 뒤집기, 회문(팰린드롬) 판별**
- **구구단, 별 찍기 (중첩 반복문)**
- **스택 / 큐 동작 순서** (push/pop 결과 묻기)

`[5,3,8,1]` → 인접 비교 후 교환 → `[3,5,1,8]` (가장 큰 값이 맨 뒤로)

### 선택 정렬 1회전 예시
`[5,3,8,1]` → 최솟값 1을 맨 앞과 교환 → `[1,3,8,5]`
