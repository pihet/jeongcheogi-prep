## 1. 프로그래밍 공통: 코드 추적과 알고리즘

1. 초기값과 자료형을 적는다. 2. 반복마다 변수 표를 한 줄씩 갱신한다. 3. 출력문을 만나는 즉시 결과에 이어 적는다. 4. 함수 호출은 매개변수와 반환값을 분리한다. 5. 마지막에 공백·개행·자료형을 검산한다.

| 패턴 | 판별법 |
|---|---|
| 순차 탐색 / 이진 탐색 | 이진 탐색은 **정렬된** 자료에서 중간값과 비교해 범위를 반씩 줄임 |
| 버블 / 선택 / 삽입 정렬 | 인접 교환 / 최솟값 선택 / 앞쪽 정렬 구간에 끼워 넣기 |

**연습:** `[5,3,8,1]`의 버블 정렬 한 회전은 `[3,5,1,8]`, 선택 정렬 한 회전은 `[1,3,8,5]`. 어떤 위치가 확정되는지까지 설명할 수 있어야 한다.


## 2. C: 값·포인터·문자열·비트

| 구분 | 핵심 |
|---|---|
| `i++` / `++i` | 현재 식에는 증가 전 값 / 증가 후 값 사용 |
| 정수 나눗셈 | `7/2`는 3, `7/2.0`은 3.5. 음수 정수 나눗셈은 0쪽으로 버림 |
| 배열·포인터 | `a[i] == *(a+i)`, `*p++`는 `*(p++)`, `(*p)++`는 가리키는 값을 증가 |
| 함수 인자 | 포인터를 받아 `*p`를 바꾸면 호출자 객체 변경 가능. 일반 값 인자는 복사됨 |
| 문자열 | `char s[] = "ABC"`의 `strlen(s)=3`, `sizeof(s)=4` (`\0` 포함) |
| 구조체 | 포인터 `q->age`는 `(*q).age`와 같음 |
| `static` 지역 변수 | 호출이 끝나도 값 유지 |

비트: `&` AND, `|` OR, `^` XOR, `<<` 왼쪽 이동, `>>` 오른쪽 이동. 예: `5 & 3 = 1`, `5 ^ 3 = 6`, `5 << 2 = 20`. 출력 서식도 `%d` 정수, `%c` 문자, `%s` 문자열을 구분한다.

**주의:** `a++ + ++a`처럼 한 식에서 같은 C 변수를 순서 관계 없이 여러 번 변경하는 예는 정의되지 않은 동작일 수 있다. 이를 왼쪽부터 계산하는 암기 문제로 취급하지 않는다.

| 장면 | 결과 |

| 오버로딩 / 오버라이딩 | 같은 이름·다른 매개변수 / 상속 관계에서 메서드 재정의 |

## 4. Python: 슬라이스·별칭·출력

| 코드/개념 | 결과·주의점 |
|---|---|
| `a[1:4]`, `a[::-1]` | 끝 4는 제외 / 역순 |
| `range(1,5)`, `enumerate(xs,start=1)` | 1~4 / 번호만 1부터, 리스트 인덱스는 그대로 0부터 |
| `a.append(x)` / `a.extend(xs)` | 하나의 원소 추가 / 순회 가능한 원소들을 펼쳐 추가 |
| `a.sort()` / `sorted(a)` | 원본 정렬·반환 `None` / 새 정렬 리스트 반환 |
| `7/2`, `7//2`, `2**3` | 3.5 / 3 / 8 |
| `print(...,sep='-',end='|')` | 인자 사이 `-`, 마지막에 개행 대신 `|` |
| `dict` / `set` / `tuple` | 키-값 / 중복 없는 집합 / 변경 불가 시퀀스 |

**별칭과 얕은 복사:** `b=a`는 같은 리스트를 가리킨다. `a += [3]`은 리스트를 제자리 확장하므로 `b`에서도 보인다. `a = a + [3]`은 새 리스트를 만들어 `a`만 다시 가리킨다. `a[:]`나 `a.copy()`는 새 **바깥** 리스트지만 안쪽 중첩 객체는 여전히 공유할 수 있다. `a += 'xy'`는 문자를 각각 원소로 추가한다.

**조건부 출력:** `f'{i}{v[0].upper()}' if len(v)>3 else f'{i}{v}'`는 길이가 4 이상일 때만 첫 글자를 대문자로 만든다. `end='|'`이면 각 출력 뒤에 파이프가 남는다.


## 5. SQL ①: 종류·구조·제약조건

| 종류 | 역할 | 대표 명령 |
|---|---|---|
| DDL | 테이블 등 구조 정의 | `CREATE`, `ALTER`, `DROP`, `TRUNCATE` |
| DML | 행 데이터 조회·변경 | `SELECT`, `INSERT`, `UPDATE`, `DELETE` |
| DCL | 접근 권한 제어 | `GRANT`, `REVOKE` |
| TCL | 트랜잭션 제어 | `COMMIT`, `ROLLBACK`, `SAVEPOINT` |

교재에 따라 TCL을 DCL에 묶기도 한다. 문제의 분류 기준을 따른다. `DELETE`는 조건에 맞는 행, `TRUNCATE`는 모든 행을 빠르게 비우되 구조 유지, `DROP`은 테이블 구조 자체를 없앤다. 트랜잭션·롤백 가능 여부는 DBMS와 실행 맥락에 따라 달라지므로 절대 표현을 피한다.

```sql
CREATE TABLE 수강 (
  수강번호 CHAR(5) PRIMARY KEY,
  학번 CHAR(4) NOT NULL,
  과목명 VARCHAR(30),
  CONSTRAINT FK_수강_학생 FOREIGN KEY (학번)
    REFERENCES 학생(학번) ON DELETE CASCADE
);
ALTER TABLE 수강 DROP COLUMN 과목명;
```

`PRIMARY KEY`=유일·NULL 불가, `FOREIGN KEY`=다른 테이블 키 참조, `UNIQUE`=중복 방지, `NOT NULL`=NULL 금지, `CHECK`=조건 검사, `DEFAULT`=기본값. `CASCADE`는 참조 행의 연쇄 삭제를 뜻하는 문맥인지 확인한다.


## 6. SQL ②: SELECT·NULL·그룹·JOIN

논리적 처리 흐름은 대체로 `FROM/JOIN → WHERE → GROUP BY → HAVING → SELECT → ORDER BY`다. `WHERE`는 그룹화 **전** 행 필터, `HAVING`은 그룹화 **후** 그룹 필터.

```sql
SELECT 학과코드, COUNT(*) AS 인원, AVG(성적) AS 평균
FROM 학생
WHERE 성적 IS NOT NULL
GROUP BY 학과코드
HAVING COUNT(*) >= 2
ORDER BY 평균 DESC;
```

`WHERE 성적 IS NOT NULL`이면 그 행은 **COUNT(*)에서도 제외**된다. WHERE가 없다면 `COUNT(*)`는 성적 NULL인 행도 세지만 `COUNT(성적)`과 `AVG(성적)`은 성적 NULL을 빼고 계산한다. `GROUP BY 학과코드`에서 학과코드 NULL 행들은 하나의 NULL 그룹을 이룰 수 있다. `AVG`에 자동 소수 자릿수 규칙은 없다. 요구할 때 `ROUND(AVG(성적),2)`처럼 명시한다.

| 문법 | 뜻 |
|---|---|
| `IS NULL` / `IS NOT NULL` | NULL 비교 (`= NULL` 아님) |
| `LIKE '김%'` / `LIKE '_길동'` | 김으로 시작 / 한 글자 뒤에 길동 |
| `BETWEEN a AND b` | 양 끝 포함 |
| `IN (...)` / `EXISTS (...)` | 목록에 속함 / 결과 행 존재 |
| `ORDER BY x DESC` | 내림차순, 생략 시 ASC |

`INNER JOIN`은 양쪽이 일치하는 행, `LEFT JOIN`은 왼쪽의 모든 행과 오른쪽의 일치 행을 남긴다. 매칭이 없으면 오른쪽 열에 NULL이 온다. `CROSS JOIN`은 모든 조합, `SELF JOIN`은 같은 테이블을 별칭 두 개로 조인한다. `UNION`은 중복 제거, `UNION ALL`은 중복 유지. `> ALL`은 모든 결과보다 커야 하고 `> ANY`는 하나보다만 커도 된다.


## 7. SQL ③: 권한·뷰·인덱스·트리거·함수

```sql
GRANT SELECT ON 학생 TO 사용자A;
REVOKE SELECT ON 학생 FROM 사용자A;
CREATE VIEW 우수학생 AS SELECT 학번, 이름 FROM 학생 WHERE 성적 >= 90;
CREATE INDEX idx_학생_이름 ON 학생(이름);
```

`GRANT ... TO`와 `REVOKE ... FROM`을 구분한다. 뷰는 저장된 쿼리로 정의하는 가상 테이블이며, **단순 뷰는 갱신 가능할 수도 있으므로** '모든 뷰는 수정 불가'로 외우지 않는다. 인덱스는 검색에 유리할 수 있지만 쓰기·저장 비용이 있다.

트리거 골격: `CREATE TRIGGER 이름 BEFORE|AFTER INSERT|UPDATE|DELETE ON 테이블 ...` → **시점·이벤트·대상·동작**을 구별한다. `FOR EACH ROW`는 행별 트리거를 뜻한다. `:NEW` 같은 문법과 프로시저 매개변수·호출 방식은 DBMS별로 다르므로 문제에 지정된 방언을 확인한다.

함수: `COUNT`, `AVG`, `SUM`, `MAX`, `MIN`; 문자열 `UPPER`, `LOWER`, `SUBSTR`, `LENGTH`, `TRIM`; NULL 대체 `COALESCE`(표준), `NVL`(Oracle); 반올림 `ROUND`. 날짜·문자 변환 함수는 DBMS별 차이가 크다.


## 8. DB 이론: 설계·관계·정규화·트랜잭션

**설계 순서:** 요구사항 분석 → 개념적 설계(**E-R 다이어그램**) → 논리적 설계(테이블·정규화) → 물리적 설계(인덱스·저장 구조) → 구현. 외부 스키마=사용자 관점, 개념 스키마=DB 전체 논리 구조, 내부 스키마=물리 저장 구조.

| 용어 | 뜻 |
|---|---|
| 릴레이션 / 튜플 / 속성 | 테이블 / 행(레코드) / 열(필드) |
| 도메인 | 한 속성이 가질 수 있는 값의 범위 |
| 카디널리티 / 차수 | 행 개수 / 열 개수 (관계 모델 문맥) |
| 슈퍼키 / 후보키 | 유일성을 만족하는 키 / 유일성과 최소성을 모두 만족하는 키 |
| 기본키 / 대체키 / 외래키 | 선택된 후보키 / 선택되지 않은 후보키 / 다른 관계의 키를 참조하는 속성 |

**정규화:** 1NF 원자값 → 2NF 부분 함수 종속 제거 → 3NF 이행 함수 종속 제거 → BCNF 모든 결정자가 후보키 → 4NF 다치 종속 제거 → 5NF 조인 종속 제거. 삽입·삭제·갱신 이상을 줄이는 목적이다. 반정규화는 성능 등을 위해 의도적으로 중복을 허용한다.

**ACID:** 원자성(전부 또는 전무), 일관성(제약 유지), 격리성(동시 트랜잭션 간 간섭 제어), 영속성(커밋 결과 지속). 병행 제어는 로킹·타임스탬프·낙관적 검증·MVCC, 회복은 로그·UNDO/REDO·검사점·그림자 페이징이다. **검사점은 병행 제어가 아니라 회복**이다.

관계대수: 순수 관계 연산은 선택(SELECT)·투영(PROJECT)·조인(JOIN)·나눗셈(DIVISION), 일반 집합 연산은 합·교·차·카티션 곱이다.


## 9. 네트워크: 계층·주소·라우팅·HDLC

OSI 계층은 물리(비트) → 데이터링크(프레임) → 네트워크(패킷) → 전송(TCP 세그먼트/UDP 데이터그램) → 세션 → 표현 → 응용. 스위치는 보통 2계층, 라우터는 3계층이다.

| 용어 | 한 줄 판별 |
|---|---|
| TCP / UDP | 연결형·순서/재전송 보장 / 비연결형·그 보장 없음 |
| ARP / DNS / DHCP / NAT | IP→MAC / 도메인명→IP / IP 설정 자동 할당 / 사설↔공인 주소 변환 |
| ICMP / IGMP / SNMP | 오류·진단 / 멀티캐스트 그룹 / 네트워크 장비 관리 |
| RIP / OSPF / BGP | 거리 벡터 IGP / 링크 상태·SPF IGP / 경로 벡터 EGP |
| HDLC I / S / U | 사용자 정보 / 흐름·오류 제어(RR·RNR·REJ) / 링크 관리 |
| HDLC NRM / ARM / ABM | 주국 허가 필요 / 불균형에서 종국 자율 송신 / 균형에서 양쪽 자유 송신 |

TCP 연결 3단계는 SYN → SYN+ACK → ACK. IPv4는 32비트, IPv6는 128비트. IPv4에서 클래스 표기는 기본 개념이고, 실제 서브넷 계산은 **주어진 CIDR/마스크**를 기준으로 한다.


## 10. 서브네팅: 식과 경계로 풀기

1. 마스크의 1 개수를 세어 `/n`을 구한다. 2. **바뀌는 옥텟**에서 블록 크기 `256 − 마스크 옥텟`을 구한다. 3. IP가 속한 블록의 시작이 네트워크 주소다. 4. 다음 블록 시작의 직전 주소가 브로드캐스트다.

`255.255.254.0` = `/23`: 세 번째 옥텟의 블록 크기 2. `192.168.11.20`은 `10~11` 구간 → 네트워크 `192.168.10.0/23`, 브로드캐스트 `192.168.11.255`. `255.255.248.0` = `/21`: 세 번째 옥텟의 블록 크기 8. `172.16.35.7`은 `32~39` 구간 → 네트워크 `172.16.32.0/21`.

호스트 비트는 `32−n`. 전통적인 일반 서브넷에서 사용 가능한 호스트 수는 `2^(32−n)−2`다(네트워크·브로드캐스트 주소 제외). **/31 점대점 링크와 /32 단일 호스트는 별도 용법**이 있으므로 기계적으로 `−2`를 적용하지 않는다. 필요한 호스트 N대이면 `2^h−2 ≥ N`인 최소 h를 찾고 프리픽스 `/32−h`를 택한다.

암산: `/23`=세 번째 옥텟 2개씩, `/22`=4개씩, `/21`=8개씩, `/20`=16개씩. `/25` 이상이면 네 번째 옥텟의 경계를 본다.


## 11. 운영체제: 스케줄링·메모리·교착·저장

| 주제 | 구분 |
|---|---|
| 비선점 / 선점 | FCFS·SJF·HRN / RR·SRT·다단계 피드백 큐. 우선순위 방식은 구현에 따라 둘 다 가능 |
| RR 퀀텀 | 한 번에 연속 사용할 수 있는 최대 시간. 작업이 남은 채 만료되면 준비 큐로 |
| HRN | `(대기 시간 + 서비스 시간) / 서비스 시간`, 큰 값 우선 |
| 페이지 교체 | FIFO(먼저 들어온 것), LRU(가장 오래 미사용), LFU(최소 사용 빈도), OPT(미래 기준) |
| 벨레이디의 모순 | 프레임 증가에도 페이지 부재 증가 가능; FIFO의 대표 함정 |
| 스래싱 / 워킹 셋 | 페이지 교체 과다로 실행 저하 / 일정 기간 자주 참조한 페이지 집합 |
| 교착 4조건 | 상호 배제·점유와 대기·**비선점**·환형 대기 |
| RAID 0 / 1 / 5 / 10 | 스트라이핑·무중복 / 미러링 / 분산 패리티 / 미러링 쌍의 스트라이핑 |

교착 해결은 예방(조건 하나 깨기), 회피(안전 상태 유지·은행원 알고리즘), 발견, 회복을 구분한다. 메모리 배치의 최초 적합·최적 적합·최악 적합도 비교한다. RR 평균 대기 시간 계산은 간트 차트를 그려 `반환=완료−도착`, `대기=반환−실행`으로 구한다.

리눅스: `chmod` 권한 변경, `chown` 소유자 변경, `ps` 프로세스 확인, `kill` 신호 전달, `grep` 텍스트 검색. `chmod 755`는 소유자 `rwx`, 그룹·기타 `r-x`.


## 12. 보안: 목표·암호·공격·방어

**CIA:** 기밀성(허가된 사람만 접근), 무결성(변조 방지·탐지), 가용성(필요할 때 사용). 인증은 신원 확인, 인가는 권한 확인, 부인 방지는 행위 부인을 어렵게 함.

| 구분 | 예시·핵심 |
|---|---|
| 대칭키 | AES·SEED·ARIA: 같은 비밀키, 빠르지만 키 공유 문제 |
| 공개키 | RSA·ECC: 공개키/개인키 쌍, 서명·키 교환/암호화 용도 구분 |
| 해시 | SHA-256 등: 일방향 요약값. MD5·SHA-1은 보안용 신규 설계에 부적절 |
| DAC / MAC / RBAC | 소유자 재량 / 보안 등급 / 역할 기준 접근 제어 |
| WEP / WPA-TKIP / WPA2-CCMP / WPA3 | 약한 초기 방식 / 과도기적 개선 / AES 기반 / 후속 방식 |

**공격 구분:** SQL 인젝션=입력으로 SQL 구조 변경, XSS=악성 스크립트가 사용자 브라우저에서 실행, CSRF=인증된 사용자 권한으로 원치 않는 요청, 스니핑=도청, 스푸핑=신분 위장, 랜섬웨어=암호화 후 금전 요구, APT=표적 장기 공격, 워터링 홀=표적 집단이 방문하는 사이트 감염, 스피어 피싱=특정 개인 대상 피싱, 파밍=주소 해석 등을 조작해 가짜 사이트로 유도, 스미싱=문자 이용 피싱.

**방어:** 매개변수화된 쿼리·입력 검증(SQLi), 출력 인코딩·CSP(XSS), CSRF 토큰·SameSite 쿠키(CSRF), 최소 권한·다중 인증·로그 모니터링. 방화벽=트래픽 정책, IDS=탐지, IPS=차단, WAF=웹 요청 방어. 레이스 컨디션/TOCTOU는 검사 시점과 사용 시점 사이 상태가 달라지는 틈을 노린다.


## 13. 소프트웨어 공학: 요구·모델·품질·패턴

**생명주기:** 폭포수(순차), 프로토타입(시제품 피드백), 나선형(계획→위험 분석→개발→평가 반복), 애자일(짧은 반복·변화 대응). XP의 가치와 실천(Pair Programming·TDD·리팩터링·CI)을 구분한다. 스크럼은 제품 백로그→스프린트 계획→수행→검토→회고로 흐른다.

**요구사항:** 도출→분석→명세→확인. 기능적=무엇을 할지(로그인), 비기능적=어떻게 동작할지(성능·보안·운영환경·자원). '윈도우/리눅스 지원'은 운영환경, '메모리 8GB 필요'는 자원. DFD는 프로세스 원·자료 흐름 화살표·저장소 평행선·외부 개체 사각형.

**UML:** 정적 구조=클래스·객체·컴포넌트·배치·패키지 등, 동적 행위=유스케이스·시퀀스·활동·상태 등. `include`는 필수 공통 흐름, `extend`는 조건부 확장. 객체지향은 캡슐화·정보 은닉·상속·다형성·추상화. SOLID는 단일 책임·개방폐쇄·리스코프 치환·인터페이스 분리·의존 역전.

**결합도 낮은 쪽→높은 쪽:** 자료→스탬프→제어→외부→공통→내용. 다른 모듈의 **내부를 직접 건드리면 내용 결합도**. **응집도 높은 쪽→낮은 쪽:** 기능→순차→통신→절차→시간→논리→우연.

**패턴을 설명으로 구별:** Singleton(인스턴스 하나), Factory Method(생성 책임 하위 클래스로), Adapter(인터페이스 변환), Decorator(기능을 동적으로 덧붙임), Proxy(대리 접근), Observer(상태 변화 통지), Strategy(알고리즘 교체), State(상태별 행동 변화). 이름뿐 아니라 의도를 한 문장으로 설명한다.

## 15. 애플리케이션 테스트: 원리·기법·계산

| 구분 | 빠른 판별 |
|---|---|
| 정적 / 동적 분석 | 실행 없이 코드 검사 / 실행하면서 동작 검사 |
| 화이트박스 / 블랙박스 | 내부 경로·조건 / 입출력 명세 |
| 구문 / 분기 / 조건 커버리지 | 문장 실행 / 분기 결과 / 개별 조건 참·거짓 |
| 동치 분할 / 경곗값 | 같은 동작 예상 범주 대표값 / 경계와 바로 옆 값 |
| 단위→통합→시스템→인수 | 작은 구성요소에서 사용자 요구 수용까지 |
| 하향식 Stub / 상향식 Driver | 아직 없는 하위 모듈 대역 / 아직 없는 상위 호출자 대역 |
| 알파 / 베타 | 개발사 통제 환경 / 실제 사용자 환경 |
| 회귀 테스트 | 변경 후 기존 기능의 부작용 확인 |

**테스트 원리:** 결함은 존재를 보일 수 있지만 무결함은 증명하기 어렵다. 같은 테스트만 반복하면 새 결함을 덜 찾는 **살충제 패러독스**, 결함이 일부 모듈에 몰리는 **결함 집중**, 요구를 충족하지 못하면 결함 0개여도 쓸모없는 **오류 부재의 궤변**을 구분한다.

연결된 하나의 제어 흐름 그래프에서 순환 복잡도 `V(G)=E−N+2`(간선 E, 노드 N). 연결 성분이 P개인 일반형은 `E−N+2P`다. 예: E=12, N=9이면 **5**. 식만 적지 말고 최종 숫자를 적는다.


## 16. 주변 신기술·데이터 용어: 비교 중심

| 데이터 마이닝 / 데이터 웨어하우스 | 데이터에서 패턴 탐색 / 분석용 데이터를 통합·저장 |
| 제로 트러스트 / 경계 중심 보안 | 내·외부 모두 지속 검증 / 내부망을 상대적으로 신뢰 |

블록체인=분산 원장, 디지털 트윈=현실 대상의 가상 모델, RPA=반복 업무 자동화, NoSQL=비관계형 데이터 모델 계열, MapReduce=분산 병렬 처리 모델. 새로운 용어는 **무엇을 분리·가상화·자동화하는가**를 한 문장으로 분해하면 낯선 보기도 구별하기 쉽다.


## 17. 최종 셀프 점검 — 답을 가리고 말하기

# 이해편 — 용어가 아니라 문제를 푸는 설명

앞의 1~16장은 시험 직전에 찾기 쉬운 압축표다. 여기서는 표에서 헷갈리기 쉬운 개념을 **예제 → 실제로 일어나는 일 → 정답을 가르는 단서** 순서로 다시 풀어 쓴다. 표만 읽고 막혔다면 해당 설명을 먼저 읽고 돌아가면 된다.


## 설명 A. C 포인터는 '주소가 움직였나, 값이 바뀌었나'를 본다

`int a[2]={10,20}; int *p=a;`라면 p에는 첫 원소의 주소가 들어 있다. `*p`는 그 주소의 값인 10이다. 여기서 `printf("%d", *p++);`를 실행하면 **10을 출력하고 p가 두 번째 원소로 이동**한다. `++`가 p에 붙었고, 후위 증가이기 때문이다. 이제 `*p`는 20이다. 반면 `(*p)++`는 괄호 때문에 p가 가리키는 **값 20을 21로 바꾸며**, p 주소 자체는 그대로다.

이 차이는 외워서 맞히기보다 식을 `*(p++)`와 `(*p)++`로 다시 써 보면 보인다. 포인터 문제는 매 줄마다 'p가 가리키는 위치'와 '그 위치의 값'을 별도 칸에 기록하자. 문자열도 같은 사고방식이다. `"ABC"`를 담는 배열에는 끝을 표시하는 `\0`이 하나 더 있으므로 `sizeof`는 4지만, 글자 수를 세는 `strlen`은 3이다.


## 설명 B. Java는 '변수의 타입'과 '진짜 객체'가 다를 수 있다

`A obj = new B();`를 읽으면 obj라는 변수는 **A형 참조**이고 실제로 만들어진 물체는 **B 객체**다. `obj.f()`처럼 재정의된 인스턴스 메서드를 호출할 때는 실행 중인 실제 객체 B의 메서드가 선택된다. 반대로 A와 B에 같은 이름의 필드 `x`가 각각 있으면 `obj.x`는 참조 변수의 선언 타입 A가 가진 필드를 본다. 그래서 '메서드는 객체, 필드는 변수 타입'이라고 줄여 외울 수 있지만, 두 타입을 먼저 분리해 적어야 실수하지 않는다.

문자열의 `==`도 비슷하다. `new String("abc")`와 다른 문자열 객체는 글자가 같아도 서로 다른 객체일 수 있다. 문자열 **내용**을 묻는다면 `.equals()`를 사용한다. `substring(1,3)`은 1번 위치부터 3번 직전까지만 가져오므로 두 글자를 얻는다.


## 설명 C. Python 리스트는 '같은 상자냐, 새 상자냐'로 푼다

`a=[1,2]; b=a`라고 하면 a와 b라는 이름표가 **같은 리스트 상자**에 붙는다. `a += [3]`은 그 상자 안에 3을 추가하므로 `b`로 봐도 `[1,2,3]`이다. 그런데 `a = a + [3]`은 더한 결과로 **새 리스트 상자**를 만들고 a 이름표만 옮긴다. 이때 b는 원래 리스트를 계속 가리킨다. `a.copy()`는 바깥 상자를 새로 만들지만, 리스트 안에 또 리스트가 있으면 그 안쪽 상자는 공유될 수 있다.

`enumerate(['red','green'], start=1)`은 순서쌍 `(1,'red'), (2,'green')`을 내놓는다. 리스트 자체의 인덱스를 1부터 바꾸는 기능이 아니다. 조건부 출력도 각 단어에 대해 조건을 다시 확인해야 한다. `len('red')`는 3이므로 `len(v)>3`이 거짓이고, `len('green')`은 5라 참이다. `print(...,end='|')`는 마지막 항목 뒤에도 `|`를 출력한다.


## 설명 D. SQL은 필터가 먼저냐, 계산이 먼저냐가 전부다

학생 데이터가 `(C01,90)`, `(C01,NULL)`, `(E01,80)` 세 행이라고 하자. `GROUP BY 학과코드`만 하면 C01에는 **두 행**이 들어 있다. 그래서 C01의 `COUNT(*)`는 2지만 `COUNT(성적)`은 1, `AVG(성적)`은 90이다. NULL은 그룹을 지우는 게 아니라 **해당 열을 사용하는 집계 계산에서 빠지는 것**이다.

여기에 `WHERE 성적 IS NOT NULL`을 추가하면 그룹을 만들기 전에 C01의 NULL 성적 행이 제거된다. 그러면 C01의 `COUNT(*)`도 1이 된다. 마지막으로 `HAVING COUNT(*)>=2`를 붙이면 C01과 E01 모두 조건을 통과하지 못한다. 이 순서를 머릿속으로 섞으면 '평균은 맞는데 인원이 틀린' 답이 나온다. `AVG` 결과의 소수 자릿수는 문제에서 요구한 `ROUND`나 DBMS 표시 방식에 따라 판단하고 임의로 반올림하지 않는다.

`LEFT JOIN`은 왼쪽 테이블을 끝까지 남기는 방식이다. 매칭되지 않은 오른쪽 데이터는 NULL로 채운다. `WHERE 오른쪽열 IS NOT NULL`을 나중에 걸면 그 행이 다시 제거될 수 있으므로, **조인 결과와 WHERE 필터를 순서대로** 적용해야 한다.


## 설명 E. DB 설계·키·회복은 각자 묻는 질문이 다르다

학생을 저장한다고 생각하자. 먼저 '무슨 정보를 관리하나?'를 모아 요구사항을 정리한다. 학생·학과와 둘의 관계를 그림으로 표현하면 **개념적 설계의 E-R 다이어그램**이다. 그 그림을 실제 테이블과 기본키·외래키로 옮기면 **논리적 설계**이고, 어떤 인덱스를 만들고 파일에 어떻게 배치할지는 **물리적 설계**다. 문제에서 'E-R'이 보이면 개념적, '정규화'가 보이면 논리적, '인덱스·저장 위치'가 보이면 물리적을 먼저 떠올리면 된다.

학번 혼자 학생을 유일하게 구분한다면 `{학번, 이름}`도 유일하긴 하다. 하지만 이름을 빼도 되니 최소성이 없다. 그래서 `{학번, 이름}`은 **슈퍼키이지만 후보키는 아니다**. 테이블에 행이 20개, 열이 4개면 카디널리티는 20, 차수는 4다. 영어 단어만 외우지 말고 '행 개수냐 열 개수냐'를 적어 본다.

여러 거래가 동시에 DB를 건드릴 때 순서를 안전하게 조정하는 것이 **병행 제어**다(로킹 등). 장애가 난 뒤 로그를 이용해 되돌리거나 다시 실행하는 것은 **회복**이다. 검사점은 복구를 어디서 시작할지 판단하는 기준점이므로 회복 쪽에 놓는다.


## 설명 F. 서브넷은 외운 블록의 '시작 숫자'를 찾는다

`192.168.35.77/21`에서 앞의 16비트는 첫 두 옥텟을 그대로 사용한다. `/21`은 세 번째 옥텟의 앞 5비트만 네트워크로 사용한다는 뜻이라 뒤의 3비트가 변할 수 있다. 따라서 한 묶음이 `2³=8`개이고, 세 번째 옥텟은 `0~7`, `8~15`, …, `32~39`처럼 나뉜다. 35는 `32~39`에 있으니 시작인 32를 쓰고, 네 번째 옥텟은 0으로 만들어 **`192.168.32.0/21`**이 된다. 35에서 8을 무조건 빼는 게 아니다.

라우팅도 '장비가 무슨 정보를 알고 결정하나'로 구분한다. RIP는 이웃까지 몇 홉인지 같은 거리 정보를, OSPF는 링크 상태를 모아 네트워크 지도를 만들고 최단 경로를 계산한다. BGP는 조직·자율 시스템 사이의 경로 정보를 교환한다. HDLC에서 RR·RNR·REJ는 전송할 사용자 데이터가 아니라 흐름과 오류를 **감독**하는 신호이므로 S 프레임이다.


## 설명 G. 운영체제는 '무엇을 빼앗는가'를 분리한다

선점형 스케줄링은 CPU를 쓰는 프로세스를 중간에 멈추고 다른 프로세스에게 CPU를 줄 수 있다. RR에서 퀀텀이 2이고 P1의 남은 실행이 3, P2가 2라면, 둘 다 시각 0에 준비됐을 때 `P1(0~2) → P2(2~4) → P1(4~5)`처럼 진행된다. P1이 첫 퀀텀 끝에 완성되지 않았으므로 준비 큐 뒤로 간 것이다. 하지만 처음 2 안에 작업이 끝났다면 더 이상 큐로 돌아가지 않는다.

교착상태의 **비선점**은 CPU 할당이 아니라 보유 중인 자원을 강제로 회수할 수 없다는 조건이다. '선점 스케줄링'과 같은 단어가 나와도 대상 자원이 달라서 따로 읽어야 한다. RAID 0은 데이터를 나눠 저장해 빠르지만 복제·패리티가 없어 고장 복구가 안 된다. RAID 1은 똑같이 복사하는 미러링이다. '빠르게 분산'인지 '같은 내용을 복제'인지가 정답 단서다.


## 설명 H. 보안은 피해가 '어디서 일어나는가'를 본다

공격자가 검색창에 SQL 조각을 넣어 원래 쿼리의 구조를 바꾸려 하면 **SQL 인젝션**이다. 게시글에 악성 스크립트를 넣어 다른 방문자의 브라우저에서 실행되게 하면 **XSS**다. 이미 로그인한 사용자의 브라우저를 이용해 사용자가 원치 않는 송금 요청을 보내게 하면 **CSRF**다. 세 공격 모두 웹에서 일어나지만, 각각 DB 쿼리·방문자 브라우저·인증된 요청이 핵심 대상이다.

특정 회사 직원들이 자주 가는 사이트를 먼저 감염시켜 기다리는 것은 **워터링 홀**이다. 특정 직원에게 맞춤형 가짜 메일을 보내는 것은 **스피어 피싱**이고, 정확한 주소를 입력해도 주소 해석을 속여 가짜 사이트로 보내면 **파밍**이다. '대상 집단이 평소 방문하는 장소'가 단서인지, '속이는 메시지'인지, '주소를 바꿔치기했는지'를 구별하자.


## 설명 I. 소프트웨어 공학은 작업의 결과물을 떠올린다

사용자에게 '무엇이 필요하냐'고 듣는 것은 요구사항 **도출**이다. 서로 충돌하거나 모호한 요구를 조정하면 **분석**, 개발자가 구현할 수 있게 문서로 명확히 남기면 **명세**, 그 문서가 실제 사용자 요구를 맞게 표현했는지 검토하면 **확인**이다. 그래서 순서만 외우는 것보다 각 단계의 결과물을 머리에 그리면 순서가 자연스럽다.

모듈 A가 B에게 필요한 값 하나만 넘기면 결합도가 낮다. A가 B의 내부 변수에 직접 들어가 값을 고쳐야만 동작하면 B 내부 구현이 바뀔 때 A도 깨진다. 이것이 가장 나쁜 **내용 결합도**다. 디자인 패턴도 이름보다 상황을 본다. 여러 화면에 상태 변경을 알려야 하면 Observer, 실행할 알고리즘을 상황에 따라 바꾸면 Strategy, 인터페이스가 맞지 않는 두 부품을 연결하면 Adapter다.


## 설명 J. 인터페이스·테스트·배포는 한 흐름으로 연결된다

댓글 화면을 만든다고 하자. 화면이 서버에 댓글 목록을 요청하고 일부만 새로 그리는 것은 **AJAX** 방식이다. 그 요청이 `/comments/42` 같은 자원을 대상으로 설계되는 것은 **REST** 스타일일 수 있다. 실시간으로 새 댓글이 도착하자마자 양방향 연결로 받는다면 **WebSocket**을 쓸 수 있다. 세 단어는 같은 기능에서 함께 쓰일 수도 있으니 무조건 배타적인 보기로 외우지 않는다.

기능을 만든 뒤에는 함수 하나를 단위 테스트하고, 다른 모듈과 연결해 통합 테스트하고, 전체 시스템을 확인한 뒤 사용자가 인수 테스트한다. 상위 모듈부터 내려가면 아직 없는 하위 모듈의 흉내인 **Stub**, 하위부터 올라가면 호출해 줄 상위 모듈의 흉내인 **Driver**가 필요하다. 코드를 바꾼 뒤 이전 기능이 여전히 되는지 다시 확인하면 **회귀 테스트**다.

검증된 코드는 버전을 붙여 빌드하고 설치 파일·의존성·사용 설명을 함께 묶어 배포한다. 형상 관리의 통제는 변경을 **승인**하는 것이고 감사는 실제 반영 결과가 **맞는지 점검**하는 것이다. 두 단계가 같은 뜻이 아니다.


## 설명 K. 신기술은 앞부분이 비슷해도 '대상'이 다르다

**SDN**은 네트워크 장비마다 흩어진 제어 결정을 소프트웨어 제어부로 모으는 개념이다. 데이터가 실제로 지나가는 전달부와 '어디로 보낼지'를 결정하는 제어부를 분리한다. **NFV**는 방화벽·라우터 같은 네트워크 기능을 전용 장비 대신 범용 컴퓨터의 소프트웨어로 구현한다. 둘 다 네트워크를 소프트웨어로 다루지만 SDN은 **제어 방식**, NFV는 **기능의 구현 장소**가 핵심이다.

**MSA**는 주문·결제·배송을 독립 서비스로 나눠 각각 배포할 수 있게 한다. 그렇다고 무조건 더 좋은 것은 아니다. 서비스끼리 통신하고 장애를 추적하는 비용이 늘 수 있다. **엣지 컴퓨팅**은 카메라처럼 데이터가 생기는 곳 가까이서 처리해 지연·전송량을 줄이고, 클라우드는 중앙에서 필요한 계산 자원을 유연하게 빌려 쓰는 쪽이다. 새로운 용어도 '무슨 문제를 해결하려고 무엇을 분리·옮겼나'로 풀면 외울 양이 줄어든다.


## 설명 L. 연계와 UI는 '누가 누구와 대화하나'로 나눈다

회사에 주문 시스템과 재고 시스템이 따로 있다고 하자. 주문이 들어오면 재고를 차감해야 하므로 두 프로그램이 데이터를 주고받는 **연계**가 필요하다. 주문 시스템이 재고 시스템의 API를 즉시 호출하면 상대 응답을 기다리는 동기식 흐름이 된다. 메시지 큐에 '주문 발생'을 넣고 재고 시스템이 나중에 가져가면 비동기식 흐름이 된다. 큐는 잠시 메시지를 보관해 두 시스템의 속도 차이를 완충하지만, 중복 전달 가능성을 고려해 같은 주문을 두 번 처리하지 않도록 설계해야 한다.

**EAI**는 기업 내부 여러 애플리케이션을 연계하는 전체 접근이고, **ESB**는 서비스들이 공통 버스를 통해 연결되는 방식이다. JSON과 XML은 이때 오가는 내용을 표현하는 형식이다. 사용자가 명령어를 직접 치면 CLI, 버튼·아이콘을 클릭하면 GUI, 음성이나 몸짓으로 조작하면 NUI다. 문제에서 '연결 구조'를 묻는지 '메시지 형식'을 묻는지 '사용자 조작 방식'을 묻는지 먼저 구별하자.


## 설명 M. 테스트 기법은 '안쪽을 보나, 입출력만 보나'를 묻는다

나이가 18세 이상일 때 가입을 허용하는 함수를 테스트한다고 하자. 내부의 `if (age >= 18)` 코드를 보고 참·거짓 분기가 모두 실행되게 17과 18을 고르면 **화이트박스 분기 커버리지**를 의식한 설계다. 내부 코드는 모르고 요구사항만 보고 17·18·19 같은 경계 값을 골라도 **블랙박스 경곗값 분석**이 된다. 즉 같은 입력이 나와도 무엇을 근거로 케이스를 만들었는지가 다르다. 동치 분할이라면 '18 미만'과 '18 이상'의 두 범주에서 각각 대표값을 택한다.

반복문과 조건문이 늘어나면 가능한 독립 경로도 늘어난다. McCabe 순환 복잡도는 그런 독립 경로의 수를 나타내는 지표다. 연결된 제어 흐름 그래프에서 간선 E, 노드 N이면 `E−N+2`로 계산한다. 문제에 노드 9개, 간선 12개라고 적혀 있으면 식을 적고 **5**라는 최종 답까지 써야 한다. 수식만 알고 답안을 끝내지 않는 습관이 중요하다.
