## 1. 네트워크

### OSI 7계층 (물-데-네-전-세-표-응)

| 6 | 표현(Presentation) | 메시지 | JPEG, MPEG, ASCII (암호화/압축) |

| 4 | 전송(Transport) | 세그먼트 | **TCP, UDP** / 게이트웨이 |

### HDLC (비트 위주 프로토콜) ★26년1회 출제

**프레임 3종류**

| 프레임 | 이름 | 역할 |
|---|---|---|
| **I** | 정보(Information) | 사용자 데이터 전달 |
| **S** | 감독(Supervisory) | 오류 제어 · 흐름 제어 (RR, RNR, REJ, SREJ) |
| **U** | 비번호(Unnumbered) | 링크 동작 모드 설정 · 관리 |

**전송 모드 3종류**

| 모드 | 링크 | 설명 |
|---|---|---|
| **NRM** (정규 응답 모드) | 불균형 | 종국은 주국의 **허가(Poll)를 받아야만** 전송 |
| **ABM** (비동기 균형 모드) | 균형 (P2P) | 혼합국끼리 **허가 없이 언제나** 전송 |
| **ARM** (비동기 응답 모드) | 불균형 (전이중) | 종국이 **허가 없이 송신**은 가능하나, 링크 제어는 주국만 |

> 26년 1회에 프레임 종류와 전송 모드가 빈칸 ①~⑤로 출제됐다.
> **N**RM = 허가 필요, **A**BM = 균형·자유, **A**RM = 자유롭게 보내되 제어권은 없음.
> HDLC 프레임 구조는 `플래그 - 주소 - 제어 - 데이터 - FCS - 플래그` 순이다.

## 1-1. 서브네팅 계산 ★26년1회 출제

> **이건 외우는 게 아니라 손으로 푸는 것이다.** 반드시 직접 계산해보고 갈 것.

### 3단계로 끝난다

1. **서브넷 마스크를 2진수로 바꿔 1의 개수를 센다** → 그게 `/n` (프리픽스)
2. **IP와 마스크를 AND 연산**한다 → 네트워크 주소
3. **호스트 비트 = 32 - n**, 할당 가능 호스트 수 = **2^(32-n) - 2**

> `-2` 를 하는 이유 : 맨 앞은 **네트워크 주소**, 맨 뒤는 **브로드캐스트 주소**라 쓸 수 없다.

### 옥텟 값 ↔ 비트 수 (외워두면 빨라진다)

| 값 | 2진수 | 1의 개수 |
|---|---|---|
| 128 | 10000000 | 1 |
| 192 | 11000000 | 2 |
| 224 | 11100000 | 3 |
| 240 | 11110000 | 4 |
| 248 | 11111000 | 5 |
| 252 | 11111100 | 6 |
| 254 | 11111110 | 7 |
| 255 | 11111111 | 8 |

### 예제 1 — 26년 1회 실제 출제

> 호스트 A `192.168.11.20`, 호스트 B `192.168.12.200`, 서브넷 마스크 `255.255.254.0`
> A와 B의 네트워크 주소를 CIDR로 쓰시오.

`255.255.254.0` → 8 + 8 + **7** + 0 = **/23**
세 번째 옥텟만 계산하면 된다. 254 = `11111110` 이므로 **마지막 1비트를 0으로 만든다.**

- A : 11 = `00001011` → `00001010` = 10 → **192.168.10.0/23**
- B : 12 = `00001100` → `00001100` = 12 → **192.168.12.0/23**

**A와 B는 서로 다른 네트워크다.** 이걸 묻는 문제였다.

- A 네트워크의 브로드캐스트 : `192.168.11.255`
- 호스트 범위 : `192.168.10.1` ~ `192.168.11.254`
- 할당 가능 호스트 수 : 2^(32-23) - 2 = 512 - 2 = **510개**

### 예제 2 — /26

> `200.1.1.100`, 마스크 `255.255.255.192`

192 = `11000000` → /26. 네 번째 옥텟만 보면 된다.
100 = `01100100` 에서 **앞 2비트만 남기고 나머지를 0으로** → `01000000` = 64

- 네트워크 주소 : **200.1.1.64/26**
- 브로드캐스트 : `200.1.1.127`
- 호스트 범위 : `200.1.1.65` ~ `200.1.1.126`
- 호스트 수 : 2^6 - 2 = **62개**

> 요령 — /26은 블록 크기가 **64**다. 그래서 경계가 0, 64, 128, 192 로 떨어진다.
> 100은 64와 128 사이이므로 네트워크는 **64**.

### 예제 3 — 옥텟을 넘어가는 경우

> `172.16.35.7`, 마스크 `255.255.248.0`

248 = `11111000` → 8+8+5 = **/21**. 블록 크기는 256 - 248 = **8**
세 번째 옥텟 35는 32와 40 사이 → 네트워크는 **32**

- 네트워크 주소 : **172.16.32.0/21**
- 브로드캐스트 : `172.16.39.255`
- 호스트 수 : 2^11 - 2 = **2046개**

### 블록 크기로 푸는 요령

**블록 크기 = 256 − (해당 옥텟의 마스크 값)**

경계는 0부터 블록 크기씩 올라간다. IP가 속한 구간의 **시작값**이 네트워크 주소다.

| 마스크 옥텟 | 블록 크기 | 경계 |
|---|---|---|
| 192 | 64 | 0, 64, 128, 192 |
| 224 | 32 | 0, 32, 64, 96, … |
| 240 | 16 | 0, 16, 32, 48, … |
| 248 | 8 | 0, 8, 16, 24, … |
| 254 | 2 | 0, 2, 4, 6, … |

### 자주 나오는 변형

- **"N개의 서브넷으로 나누려면?"** → 2^k ≥ N 인 최소 k만큼 **마스크 비트를 늘린다.**
  (클래스 C를 4개로 → k=2 → /24 + 2 = **/26**)
- **"각 서브넷에 호스트 N대가 필요하면?"** → 2^h - 2 ≥ N 인 최소 h만큼 **호스트 비트를 남긴다.**
  (호스트 50대 필요 → 2^6-2 = 62 ≥ 50 → h=6 → **/26**)

## 2. 운영체제

### 프로세스 스케줄링
| 구분 | 알고리즘 |
|---|---|
| **비선점** | FCFS, SJF, HRN, 우선순위, 기한부(Deadline) |
| **선점** | RR(라운드 로빈), SRT, 다단계 큐, 다단계 피드백 큐(MLFQ) |

- **HRN 우선순위 = (대기시간 + 서비스시간) / 서비스시간** (값이 클수록 우선)

### 페이지 교체 알고리즘
- **FIFO**: 먼저 들어온 것 교체 (Belady 모순 발생)
- **LRU**: 가장 오래 사용 안 된 것 (Least Recently Used)
- **LFU**: 사용 빈도가 가장 적은 것 (Least Frequently Used)
- **NUR**: 최근에 사용하지 않은 것 (참조 비트 / 변형 비트)
- **OPT**: 앞으로 가장 오래 안 쓸 것 (이론상 최적)

### 교착상태(Deadlock)
- **발생 4조건: 상호배제 / 점유와 대기 / 비선점 / 환형대기**
- 해결: 예방(Prevention), 회피(Avoidance — **은행원 알고리즘**), 발견(Detection), 회복(Recovery)

### 기타
- **스래싱(Thrashing)**: 페이지 교체가 너무 잦아 처리 시간보다 페이지 교체 시간이 커지는 현상
- **워킹셋(Working Set)**: 프로세스가 일정 시간 동안 자주 참조하는 페이지 집합
- **구역성(Locality)**: 시간 구역성 / 공간 구역성
- 메모리 배치 전략: **최초 적합(First Fit), 최적 적합(Best Fit), 최악 적합(Worst Fit)**

### UNIX / LINUX 명령어
`chmod`(권한), `chown`(소유자), `ls`, `cat`, `grep`, `ps`, `kill`, `fork`(프로세스 생성)
- chmod 숫자: 읽기 4, 쓰기 2, 실행 1 → `chmod 755` = rwxr-xr-x

### RAID 레벨 ★기출 반복

| 레벨 | 이름 | 특징 |
|---|---|---|
| **RAID 0** | 스트라이핑 | 데이터를 여러 디스크에 **분산 저장**, 패리티(오류 검출) 없음, 속도는 빠르지만 **장애 시 복구 불가** |
| **RAID 1** | 미러링 | 같은 데이터를 **두 디스크에 복제**, 안정성 높지만 저장 효율 50% |
| **RAID 5** | 분산 패리티 | 데이터와 **패리티 정보를 여러 디스크에 분산 저장**, 디스크 1개 장애까지 복구 가능 |
| **RAID 1+0(10)** | 미러링+스트라이핑 | 미러링한 디스크 쌍을 다시 스트라이핑 — 성능과 안정성 모두 확보 |

> **"패리티 없음 + 스트라이핑 + 장애 시 복구 불가"** 라는 설명이 나오면 **RAID 0**이다.

### 라운드 로빈(RR) 평균 대기시간 계산 예제

> 프로세스 P1(버스트 5), P2(버스트 3), P3(버스트 4)가 모두 시각 0에 도착,
> **시간 할당량(Quantum) = 2**, 문맥 전환 시간은 무시.

**간트 차트**: P1(0~2) → P2(2~4) → P3(4~6) → P1(6~8) → P2(8~9, 완료) → P3(9~11, 완료) → P1(11~12, 완료)

| 프로세스 | 완료 시각 | 반환시간(완료−도착) | 대기시간(반환−버스트) |
|---|---|---|---|
| P1 | 12 | 12 | 12−5 = **7** |
| P2 | 9 | 9 | 9−3 = **6** |
| P3 | 11 | 11 | 11−4 = **7** |

**평균 대기시간 = (7+6+7) / 3 ≈ 6.67ms**

> RR은 **할당량이 끝나면 무조건 큐 맨 뒤로 보낸다.** 남은 시간이 할당량보다 적으면
> 그만큼만 쓰고 바로 끝난다(P2·P3의 마지막 구간처럼). 큐에 **누가 먼저 들어왔는지** 순서를 놓치면 전부 어긋난다.

## 3. 보안

### 보안 3요소 + 확장
**기밀성(Confidentiality) / 무결성(Integrity) / 가용성(Availability)**
+ 인증(Authentication), 부인 방지(Non-repudiation)

### 암호 알고리즘
| 분류 | 알고리즘 |
|---|---|
| **대칭키(비밀키)** | DES, 3DES, **AES**, **SEED**, **ARIA**, IDEA, LFSR, RC4, RC5 |
| **비대칭키(공개키)** | **RSA**, **ECC**, ElGamal, Diffie-Hellman, DSA |
| **해시(일방향)** | **MD5**, **SHA-1 / SHA-256 / SHA-512**, HAVAL, N-NASH, SNEFRU |

- 대칭키: 빠름, 키 분배 어려움, 키 개수 **n(n-1)/2**
- 비대칭키: 느림, 키 분배 쉬움, 키 개수 **2n**
- SEED / ARIA: **국내 개발** 대칭키 알고리즘

### 주요 공격 기법
| 용어 | 설명 |
|---|---|
| SQL Injection | 입력값에 SQL문을 삽입해 DB 조작 |
| **XSS** (Cross Site Scripting) | 웹페이지에 악성 스크립트 삽입 |
| **CSRF** (Cross Site Request Forgery) | 사용자 권한을 도용해 원치 않는 요청 전송 |
| 스니핑(Sniffing) | 네트워크 패킷 도청 (수동적 공격) |
| 스푸핑(Spoofing) | 신분 위장 (IP / ARP / DNS) |
| 세션 하이재킹 | 인증된 세션을 가로챔 |
| 버퍼 오버플로우 | 할당 메모리 초과 입력으로 실행 흐름 조작 |
| **APT** | 특정 대상을 장기간 지속 공격 |
| **Zero-Day** | 취약점이 공개되기 전에 이루어지는 공격 |
| 랜섬웨어 | 파일을 암호화하고 금전 요구 |
| 피싱 / 스미싱 / 파밍 | 이메일 / 문자(SMS) / DNS 변조로 유도 |
| 백도어 | 정상 인증을 우회하는 통로 |
| **Rootkit** | 침입 후 흔적 삭제 및 은닉 도구 모음 |

### 서비스 거부(DoS) 공격
- **SYN Flooding**: 연결 요청만 대량 전송해 대기 큐 고갈
- **Smurfing**: ICMP 브로드캐스트로 트래픽 증폭
- **Ping of Death**: 규격을 초과한 ICMP 패킷 전송
- **Land Attack**: 출발지 IP와 목적지 IP를 동일하게 설정
- **Teardrop**: 단편화 offset 값 조작
- **DDoS**: 분산 서비스 거부 (좀비 PC 이용)

### 접근 통제
- **DAC** (임의적, 소유자가 부여) / **MAC** (강제적, 보안 등급 기반) / **RBAC** (역할 기반)

### 보안 솔루션
방화벽(Firewall), **IDS**(침입 탐지), **IPS**(침입 방지), **VPN**, DMZ, **SIEM**, ESM, NAC, DLP

### 추가 공격 기법 ★26년1회 출제

| 용어 | 설명 | 구분 포인트 |
|---|---|---|
| **Watering Hole** | 대상 조직이 **자주 방문하는 사이트**를 미리 감염시켜 두고 방문자를 노림 | **특정 집단**을 노린다 |
| **Phishing** | 위장 메일·사이트로 정보를 빼냄 | 불특정 다수 |
| **Spear Phishing** | 특정 인물을 겨냥한 맞춤형 피싱 | 특정 개인 |
| **Pharming** | **DNS를 변조**해 정확한 주소를 입력해도 가짜 사이트로 | 주소는 맞는데 사이트가 가짜 |
| **Smishing** | 문자(SMS)를 이용한 피싱 | SMS |
| **Drive by Download** | 사이트 **접속만으로** 악성코드가 설치됨 | 클릭 없이 감염 |
| **Business SCAM** | 거래처를 사칭해 **송금 계좌를 바꿔치기** | 금전 탈취 |
| **Cyber Kill Chain** | 공격 단계를 7단계로 모델링한 **방어 프레임워크** | 공격이 아니라 방어 개념 |
| **Rootkit** | 침입 후 흔적을 지우고 은닉하는 도구 모음 | 침입 **후** |
| **Ransomware** | 파일을 암호화하고 금전 요구 | |

> 26년 1회에 Watering Hole이 보기 제시형으로 나왔다(22년 1회 재출제).
> **"조직·집단을 노린다"가 나오면 워터링 홀**이다.

### 레이스 컨디션과 심볼릭 링크 ★26년1회 출제

- **레이스 컨디션(Race Condition)** : 둘 이상의 프로세스가 **자원을 동시에 사용하려 경쟁**할 때,
  검사 시점과 사용 시점 사이의 **틈(TOCTOU)** 을 노리는 공격
- 공격자는 프로그램이 만들 임시 파일 이름으로 **심볼릭 링크**를 미리 만들어 두고,
  프로그램이 **엉뚱한 파일(예: /etc/passwd)에 쓰도록** 유도한다

| 구분 | 심볼릭 링크(Symbolic Link) | 하드 링크(Hard Link) |
|---|---|---|
| 가리키는 것 | **경로(이름)** | **inode(실제 데이터)** |
| 다른 파일시스템 | 가능 | 불가능 |
| 원본 삭제 시 | **깨진다** | 데이터는 남는다 |
| 디렉터리 대상 | 가능 | 불가능 |

> 리눅스 명령 : `ln -s 원본 링크` (심볼릭) / `ln 원본 링크` (하드)
> 26년 1회에 난이도 '상'으로 출제됐다.

## 4. 소프트웨어 공학

### 소프트웨어 생명주기 모델
- **폭포수**: 단계별 순차 진행, 이전 단계로 되돌아가기 어려움
- **프로토타입**: 견본품 제작 후 사용자 평가
- **나선형(Spiral)**: **계획 → 위험 분석 → 개발 → 고객 평가** 반복, 위험 최소화
- **애자일**: 반복적·점진적, 변화 대응 중심

### 애자일
- **XP 5가지 가치**: 용기, 단순성, 의사소통, 피드백, 존중

  - 역할: 제품 책임자(PO), 스크럼 마스터(SM), 개발팀
  - **번다운 차트**: 남은 작업량을 시각화한 그래프

### 요구사항

### 형상 관리 (Configuration Management)

**절차 4단계 : 식별 → 통제 → 감사 → 기록**

| 단계 | 하는 일 |
|---|---|
| **형상 식별** | 관리 대상(문서·코드)을 정하고 이름과 번호를 부여 |
| **형상 통제** | 변경 요청을 검토·승인하고 **베이스라인**에 반영 (변경통제위원회 CCB) |
| **형상 감사** | 변경이 **요구사항대로** 이루어졌는지 확인 |
| **형상 기록** | 변경 과정과 결과를 문서로 **기록·보고** |

- **베이스라인(Baseline)** : 정식으로 검토·합의되어 **더 이상 임의로 바꿀 수 없는 기준선**
- **버전 관리 도구** : CVS → SVN(중앙집중식) → **Git(분산형)**
- Git 주요 명령 : `commit`(로컬 저장) · `push`(원격 반영) · `pull`(원격→로컬) · `branch` · `merge`
- 형상 관리의 효과 : 무결성 보장, 변경 이력 추적, 이전 버전 복원, 동시 작업 충돌 방지

### 웹 기술 키워드 ★기출 반복

- **AJAX (Asynchronous JavaScript and XML)** : 페이지 전체를 새로고침하지 않고
  **필요한 데이터만 비동기로 서버에 요청**하여 화면 일부만 갱신하는 기법.
  자바스크립트와 XML(또는 JSON)을 이용한다.
- 헷갈리는 보기들
  - **REST** : 자원을 URL로 표현하는 **API 설계 방식** (통신 규약이지 화면 갱신 기법이 아님)
  - **WebSocket** : 클라이언트-서버 간 **양방향 실시간 연결**을 유지하는 프로토콜
  - **GraphQL** : 클라이언트가 필요한 데이터 구조를 직접 지정해 요청하는 **쿼리 언어**

> "페이지 일부만 비동기로 갱신" 이라는 설명이 나오면 거의 항상 **AJAX**다.

## 5. 테스트

### 테스트 원리 7가지
결함 존재 증명 / 완벽한 테스팅 불가능 / 초기 집중 / **결함 집중(파레토 법칙)** / **살충제 패러독스** / 정황 의존 / 오류-부재의 궤변

### 화이트박스 테스트 (구조 기반)
구문(문장) 커버리지, 결정(분기) 커버리지, 조건 커버리지, 조건/결정 커버리지, 변형 조건/결정(MC/DC), 다중 조건 커버리지, 기본 경로 테스트, 데이터 흐름 테스트, 루프 테스트

### 블랙박스 테스트 (명세 기반)
**동치 분할, 경곗값 분석, 원인-결과 그래프, 오류 예측, 비교 테스트, 결정 테이블, 상태 전이, 유스케이스, 분류 트리**

### 테스트 레벨 (V 모델)
**단위 → 통합 → 시스템 → 인수**
- 통합: **하향식(Stub 필요) / 상향식(Driver 필요)** / 샌드위치 / 빅뱅

### 스텁과 드라이버 ★26년1회 출제

| 구분 | **스텁 (Stub)** | **드라이버 (Driver)** |
|---|---|---|
| 어디에 쓰나 | **하향식** 통합 테스트 | **상향식** 통합 테스트 |
| 대신하는 것 | 아직 안 만든 **하위 모듈** | 아직 안 만든 **상위 모듈** |
| 하는 일 | 상위가 호출하는 기능을 **단순히 흉내만** 냄 | 하위 모듈에 **데이터를 넣고 결과를 확인** |

> 26년 1회 10번으로 출제됐다. **하**향식은 아래(하위)가 없으니 **S**tub,
> **상**향식은 위(상위)가 없으니 **D**river. 한글로 쓰라고 하면 **스텁 / 드라이버**.
>
> - **하향식** : 위에서 아래로. 상위 모듈부터 검증하므로 **설계 결함을 일찍 발견**한다.
> - **상향식** : 아래에서 위로. 하위 모듈부터 검증하므로 **디버깅이 쉽다**.
- 인수: 알파 테스트(개발자 환경) / 베타 테스트(사용자 환경)

### 기타
- **회귀 테스트(Regression)**: 수정 후 다른 부분에 영향이 없는지 확인
- **McCabe 순환 복잡도 V(G) = E - N + 2** (E = 간선 수, N = 노드 수)
- 결함 관리 절차: 결함 등록 → 검토 → 할당 → 수정 → 조치 확인 → 종료

|---|---|---|
| **JUnit** | 단위 테스트 프레임워크 | 자바용 xUnit 계열 오픈소스. `@Test` 애너테이션으로 테스트 코드 작성 |
| **정적 분석 도구** | 코드 분석 | 프로그램을 **실행하지 않고** 코딩 표준·문법 오류·잠재 결함을 검사 (pmd, cppcheck, SonarQube) |
| **동적 분석 도구** | 코드 분석 | 프로그램을 **실행하면서** 메모리 누수·보안 취약점·코드 커버리지를 검사 (Valgrind, Avalanche) |

> "실행하지 않고 검사" = **정적**, "실행하면서 검사" = **동적**. 이 한 단어 차이로 갈린다.

## 6. 데이터베이스

### 정규화
| 단계 | 조건 |
|---|---|
| 1NF | 모든 속성이 **원자값** |
| 2NF | **부분 함수 종속** 제거 (완전 함수 종속) |
| 3NF | **이행 함수 종속** 제거 |
| BCNF | 모든 **결정자가 후보키** |
| 4NF | **다치 종속** 제거 |
| 5NF | **조인 종속** 제거 |

- **이상 현상: 삽입 이상 / 삭제 이상 / 갱신 이상**
- 반정규화(De-normalization): 성능을 위해 의도적으로 정규화를 역행

### 트랜잭션 ACID
- **원자성(Atomicity)**: 전부 실행되거나 전부 취소
- **일관성(Consistency)**: 실행 전후 일관된 상태 유지
- **격리성(Isolation)**: 실행 중 다른 트랜잭션이 간섭 불가
- **영속성(Durability)**: 완료된 결과는 영구히 반영

### 키
슈퍼키(유일성 O, 최소성 X) / 후보키(유일성 O, 최소성 O) / 기본키 / 대체키 / 외래키

### 관계 대수
- **순수 관계 연산자**: SELECT, PROJECT, JOIN, DIVISION
- 일반 집합 연산자: 합집합, 교집합, 차집합, 카티션 프로덕트

### 무결성
개체 무결성(기본키는 NULL 불가), 참조 무결성(외래키), 도메인 무결성

### 스키마 3계층
외부 스키마(사용자 뷰) / 개념 스키마(전체 논리 구조) / 내부 스키마(물리적 저장 구조)

### 회복 기법
- **로그 기반**: 즉시 갱신, 지연 갱신
- **검사점(Checkpoint)**, **그림자 페이징(Shadow Paging)**
- UNDO(취소) / REDO(재실행)

### 병행 제어
**로킹(Locking)**, 2단계 로킹 규약, 타임스탬프, 낙관적 검증, 다중 버전 기법

### 데이터베이스 설계 절차 ★26년1회 출제

**요구사항 분석 → 개념적 설계 → 논리적 설계 → 물리적 설계 → 구현**

| 단계 | 하는 일 | 산출물 |
|---|---|---|
| **요구사항 분석** | 사용자 요구를 수집·분석하고 명세를 작성 | 요구사항 명세서 |
| **개념적 설계** | 현실 세계를 추상화하여 개체·관계를 정의 | **E-R 다이어그램**, 개념 스키마 |
| **논리적 설계** | 목표 DBMS에 맞는 **테이블 구조**로 변환하고 **정규화** | 논리 스키마(릴레이션) |
| **물리적 설계** | **저장 구조·인덱스·접근 경로**를 결정. 성능(반응시간) 고려 | 물리 스키마 |
| **구현** | DDL로 실제 테이블을 생성하고 데이터를 적재 | 데이터베이스 |

> 26년 1회에 보기에서 골라 순서대로 쓰는 문제로 나왔다.
> **논리와 물리를 바꿔 쓰는 실수가 가장 많다.**
> - **개념적** = E-R 다이어그램 그리기
> - **논리적** = 테이블(릴레이션) 확정 + **정규화**
> - **물리적** = **인덱스·저장 방식** 결정
>
> 보기에 '운영'과 '유지보수'가 함께 나오면 **함정**이다. 설계 절차 5단계에는 들어가지 않는다.

### 관계 데이터 모델 구성 요소 ★기출 반복

| 정식 용어 | 파일 구조 용어 | 의미 |
|---|---|---|
| **릴레이션(Relation)** | 파일(File) | 테이블 전체 |
| **튜플(Tuple)** | 레코드(Record) | 한 행 |
| **속성(Attribute)** | 필드/데이터 항목(Field) | 한 열 |
| **도메인(Domain)** | — | 속성이 가질 수 있는 값의 범위 |
| **카디널리티(Cardinality)** | — | 튜플(행)의 개수 |
| **디그리/차수(Degree)** | — | 속성(열)의 개수 |

> "데이터 항목·필드에 해당하는 요소" → **Attribute**, "레코드에 해당하는 요소" → **Tuple**

### 물리 설계 — 레코드 접근 방법

| 방법 | 설명 |
|---|---|
| **순차 접근** | 레코드를 저장된 순서대로 처음부터 읽음 |
| **색인 순차 접근(ISAM)** | **키 값과 주소 포인터를 쌍으로 둔 별도의 색인(인덱스) 구조**를 두어 검색 속도 향상 |
| **직접(해싱) 접근** | 키 값을 해시 함수에 넣어 바로 저장 위치를 계산해 접근 |

> "키 값 + 주소 포인터로 구성된 별도 구조로 검색 속도를 높인다" → **색인 순차 접근**
