# 1주차 — 이론 1회독 + 프로그래밍 감 잡기

**기간: 9/14(월) ~ 9/20(일)**
**목표: "아 이런 게 나오는구나"를 끝내는 주. 완벽히 외우려 하지 말 것.**

> 1주차의 함정: 1과목을 완벽하게 하려다 3과목에서 시간이 끝난다.
> **모르는 건 형광펜 치고 넘어간다.** 회독은 3주차에 또 한다.

- [ ] 애자일: XP(5가지 가치), 스크럼(백로그/스프린트/번다운차트), 린

## Day 2 (화) — 데이터 입출력 구현 (DB 이론)

### 이론 (100분)
- [ ] 데이터베이스 설계 순서: **요구조건분석 → 개념적 → 논리적 → 물리적 → 구현**
- [ ] 관계 대수 / 관계 해석 (순수관계연산자 4개: SELECT, PROJECT, JOIN, DIVISION)
- [ ] **이상현상 3가지: 삽입 이상 / 삭제 이상 / 갱신 이상**
- [ ] **정규화 (초빈다결조): 1NF(원자값) → 2NF(부분함수종속 제거) → 3NF(이행함수종속 제거) → BCNF(결정자=후보키) → 4NF(다치종속) → 5NF(조인종속)**
- [ ] 트랜잭션 **ACID: 원자성/일관성/격리성/영속성**
- [ ] 키 종류: 후보키, 기본키, 대체키, 슈퍼키, 외래키
- [ ] 인덱스, 뷰, 파티션(범위/해시/리스트/컴포지트)

### 프로그래밍 (50분)
- [ ] C 제어문: if / switch / for / while / do-while
- [ ] 연습: 구구단, 별 찍기, 1~100 합계를 **눈으로 추적**해서 출력값 맞히기

## Day 4 (목) — 서버 프로그램 구현 (디자인 패턴)

### 이론 (90분)
- [ ] 소프트웨어 아키텍처 패턴: 계층화, 클라이언트-서버, 파이프-필터, MVC, 마스터-슬레이브, 브로커
- [ ] **객체지향 5원칙 SOLID**
  - SRP(단일책임), OCP(개방폐쇄), LSP(리스코프치환), ISP(인터페이스분리), DIP(의존역전)
- [ ] **디자인 패턴 23가지 — 반드시 분류까지 암기**
  - **생성(5)**: Factory Method, Abstract Factory, Builder, Prototype, **Singleton**
  - **구조(7)**: Adapter, Bridge, Composite, Decorator, Facade, Flyweight, Proxy
  - **행위(11)**: Chain of Responsibility, Command, Interpreter, Iterator, Mediator, Memento, **Observer**, State, **Strategy**, Template Method, Visitor
- [ ] 결합도(낮을수록 좋음): **자료 < 스탬프 < 제어 < 외부 < 공통 < 내용**
- [ ] 응집도(높을수록 좋음): **기능 > 순차 > 통신 > 절차 > 시간 > 논리 > 우연**

### 프로그래밍 (60분)

### 이론 (90분)
- [ ] 테스트 원리 7가지 (결함 존재 증명, 완벽 테스팅 불가능, 초기 집중, 결함 집중(파레토), 살충제 패러독스, 정황 의존, 오류-부재의 궤변)
- [ ] **화이트박스 테스트**: 구문/결정/조건/조건-결정/변형조건-결정/다중조건, 기본 경로, 루프
- [ ] **블랙박스 테스트**: 동치 분할, 경곗값 분석, 원인-결과 그래프, 오류 예측, 비교, 결정 테이블, 상태 전이, 유스케이스
- [ ] 테스트 레벨: **단위 → 통합 → 시스템 → 인수** (V모델)
