## 문제은행 사용법

모든 문제은행은 **문제 → (메모용 빈 셀) → 접힌 정답 토글** 구조다.

1. 문제를 읽고 **종이에 답을 먼저 적는다.**
2. 그 다음 **👉 정답 · 해설 보기** 토글을 눌러 펼친다.
3. `03_Python_문제은행` 은 코드 셀이 실제로 실행되므로, 답을 적은 뒤 실행해 대조한다.
4. 틀린 것은 그날 안에 `체크리스트/오답노트_템플릿` 에 기록한다.
