## Day 23 (화) — 약어 폭격 ② + 운영체제

- [ ] 보안 파트: 암호 알고리즘 분류, 공격 기법 전부
- [ ] 운영체제
  - **프로세스 스케줄링**: 선점(RR, SRT, 다단계큐, MLFQ) / 비선점(FCFS, SJF, HRN, 우선순위, 기한부)
  - **HRN 우선순위 = (대기시간 + 서비스시간) / 서비스시간** ← 계산 문제 출제
  - **페이지 교체**: FIFO, LRU, LFU, NUR, OPT — **표 그려서 페이지 폴트 세는 연습 필수**
  - 교착상태 4조건: **상호배제, 점유와 대기, 비선점, 환형대기**
  - 교착상태 해결: 예방, 회피(**은행원 알고리즘**), 발견, 회복
- [ ] 기출 단답형 30문항
