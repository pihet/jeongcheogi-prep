# 3주차 — 기출 5개년 회독 + 약점 노출

**기간: 9/28(월) ~ 10/4(일)**
**목표: 기출을 풀면서 "내가 뭘 모르는지"를 전부 드러내는 주.**

> 이 시험은 **기출 재출제 비중이 매우 높다.** 최근 5개년(연 3회 = 약 15회분)을 2회독하면
> 단답형 체감 난이도가 확 떨어진다. 새 이론서를 펴지 말고 기출만 돌려라.

## 회독 규칙

1. **시간 재고 푼다.** 실제 시험처럼 20문항 / 150분.
2. 채점 후 **틀린 문제 + 찍어서 맞은 문제**를 오답노트로.
3. 해설을 읽고 이해했으면 **다음 날 아침에 다시 푼다.** (망각곡선)
4. 같은 문제를 3번 틀리면 → 별표 3개 → 4주차 최종 암기 리스트에 강제 편입.
