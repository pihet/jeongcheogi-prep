## SET 3 · 문항 3 — 디자인 패턴 · 결합도 · 응집도

**3.** 다음은 모듈의 **결합도(Coupling)** 에 관한 설명이다. ( ① ) ~ ( ③ )에 들어갈 유형을 보기에서 골라 쓰시오.

- ( ① ) 모듈 간에 필요한 데이터만 매개변수로 주고받는, 가장 바람직한 결합도
- ( ② ) 한 모듈이 다른 모듈의 내부 논리를 제어하는 플래그(제어 신호)를 넘기는 결합도
- ( ③ ) 한 모듈이 다른 모듈의 내부 기능이나 데이터를 직접 참조하는, 가장 나쁜 결합도

> **보기** : 자료 · 스탬프 · 제어 · 외부 · 공통 · 내용

<details>
<summary><b>👉 SET 3 문항 3 정답 · 해설 보기</b></summary>

**3.** ① 자료   ② 제어   ③ 내용

> **결합도 (낮은 것 → 높은 것)** : 자료 → 스탬프 → 제어 → 외부 → 공통 → 내용
> 순서를 나열하라는 문제가 자주 나온다. **낮을수록 좋다.**

</details>

## SET 3 · 문항 4 — 디자인 패턴 · 결합도 · 응집도

**4.** 다음은 모듈의 **응집도(Cohesion)** 에 관한 설명이다. ( ① ) ~ ( ③ )에 들어갈 유형을 보기에서 골라 쓰시오.

- ( ① ) 모듈 내 모든 기능이 단일한 목적을 수행하기 위해 구성된, 가장 바람직한 응집도
- ( ② ) 동일한 입력 데이터를 사용하거나 동일한 출력 데이터를 생성하는 기능을 묶은 응집도
- ( ③ ) 한 활동의 출력이 다음 활동의 입력으로 그대로 전달되는 응집도

> **보기** : 기능적 · 순차적 · 교환적(통신적) · 절차적 · 시간적 · 논리적 · 우연적

<details>
<summary><b>👉 SET 3 문항 4 정답 · 해설 보기</b></summary>

**4.** ① 기능적   ② 교환적(통신적)   ③ 순차적

> **응집도 (높은 것 → 낮은 것)** : 기능적 → 순차적 → 교환적 → 절차적 → 시간적 → 논리적 → 우연적
> **높을수록 좋다.** 26년 1회에 ①②③ 형태로 그대로 나왔다.
>
> **순차적과 절차적의 구분** — 데이터가 실제로 전달되면 **순차적**, 순서만 있고 데이터는 안 넘어가면 **절차적**이다.

</details>

## SET 7 · 문항 1 — IP 주소와 서브네팅 (계산형)

**1.** 호스트 A의 IP 주소는 `192.168.11.20`, 호스트 B의 IP 주소는 `192.168.12.200` 이고,
서브넷 마스크는 `255.255.254.0` 이다.
A와 B가 속한 네트워크의 주소를 각각 **CIDR 표기법**으로 쓰시오.

<details>
<summary><b>👉 SET 7 문항 1 정답 · 해설 보기</b></summary>

**1.** A : **192.168.10.0/23**   B : **192.168.12.0/23**

> 풀이 — `255.255.254.0` 은 1의 개수가 8+8+7 = **23개**이므로 `/23`.
> 세 번째 옥텟만 계산하면 된다. 254 = `11111110` 이므로 **마지막 비트를 0으로 만든다.**
> - A : 11 = `00001011` → `00001010` = **10** → 192.168.10.0/23
> - B : 12 = `00001100` → `00001100` = **12** → 192.168.12.0/23
>

## SET 1 · 문항 5 — 2026년 2회 유형 (계산형 집중)

**5.** 프로세스의 도착시간과 버스트 시간이 다음과 같다.
**SRT 스케줄링**을 적용할 때 **평균 대기시간**을 구하시오. (단위 ms)

| 프로세스 | 도착시간 | 버스트 시간 |
|---|---|---|
| P1 | 0 | 8 |
| P2 | 1 | 4 |
| P3 | 2 | 9 |
| P4 | 3 | 5 |

<details>
<summary><b>👉 SET 1 문항 5 정답 · 해설 보기</b></summary>

**5.** **6.5 ms**

> SRT는 **선점형**이다. 새 프로세스가 도착할 때마다 **남은 시간이 더 짧으면 뺏는다.**
>
> | 시각 | 사건 | 실행 |
> |---|---|---|
> | 0 | P1 시작 | P1 |
> | 1 | P2(4) 도착 < P1 남음(7) → **선점** | P2 |
> | 5 | P2 종료. 남은 것 P1=7, P3=9, P4=5 | P4 |
> | 10 | P4 종료 | P1 |
> | 17 | P1 종료 | P3 |
> | 26 | P3 종료 | — |
>
> **대기시간 = 완료시각 − 도착시간 − 버스트시간**
>
> | | 완료 | 도착 | 버스트 | 대기 |
> |---|---|---|---|---|
> | P1 | 17 | 0 | 8 | **9** |
> | P2 | 5 | 1 | 4 | **0** |
> | P3 | 26 | 2 | 9 | **15** |
> | P4 | 10 | 3 | 5 | **2** |
>
> (9 + 0 + 15 + 2) ÷ 4 = 26 ÷ 4 = **6.5 ms**
>
> 반드시 **간트 차트를 그리고 표로 계산**할 것. 선점 시점을 놓치면 전부 어긋난다.

</details>
