## 1. Python 별칭과 재할당
출력값을 두 줄로 적으시오.
```python
a = [1, 2]
b = a
a += [3]
print(b)
a = a + [4]
print(b, a)
```

<details><summary>정답·해설</summary>

```text
[1, 2, 3]
[1, 2, 3] [1, 2, 3, 4]
```
리스트의 `+=`는 기존 객체를 확장하므로 b도 본다. `a + [4]`는 새 리스트를 만들고 a만 재할당한다.
</details>

## 2. Python 딕셔너리 인덱스
출력값은?
```python
d = {'AB': 'apple', 'CD': 'cat'}
s = ''
for k, v in d.items():
    s += k[-1] + v[-1]
print(s[::-1])
```

<details><summary>정답·해설</summary>

**`tDeB`**. 순회 중 `AB→Be`, `CD→Dt`, 따라서 s=`BeDt`다. `[::-1]`로 전체를 역순으로 만든다. 두 번의 `[-1]`은 각각 마지막 글자 하나를 고른다.
</details>

## 3. enumerate와 슬라이스
출력값은?
```python
v = ['ab', 'cde', 'fg']
for i, s in enumerate(v[1:], start=1):
    print(i, s[-1], end='|')
```

<details><summary>정답·해설</summary>

**`1 e|2 g|`**. `v[1:]`은 `cde, fg`; `enumerate` 번호는 1과 2다. `s[-1]`은 마지막 문자, `end='|'`로 끝에도 `|`가 붙는다. `print`의 두 인자 사이에는 기본 공백이 있다.
</details>

## 4. C 이중 포인터
출력값은?
```c
#include <stdio.h>
void set(int **pp) { **pp = 9; }
int main(void) { int x=2; int *p=&x; set(&p); printf("%d", x); }
```

<details><summary>정답·해설</summary>

**9**. `pp`는 p의 주소, `*pp`는 p, `**pp`는 p가 가리키는 x다.
</details>

## 6. C 재귀의 음수 반환
다음 `f(4)`는?
```c
int f(int n) { if (n <= 0) return n; return f(n-2) + f(n-3); }
```

<details><summary>정답·해설</summary>

**-4**. `f(1)=f(-1)+f(-2)=-3`, `f(2)=f(0)+f(-1)=-1`, `f(4)=f(2)+f(1)=-4`. 종료 조건은 음수 인자를 0으로 바꾸지 않는다.
</details>

## 7. C static 지역 변수
출력값은?
```c
#include <stdio.h>
int f(void) { static int n=1; n*=2; return n; }
int main(void) { printf("%d ", f()); printf("%d", f()); }
```

<details><summary>정답·해설</summary>

**`2 4`**. `static` 지역 변수 n은 첫 호출 후에도 2로 남아, 두 번째 호출에서 4가 된다. 출력 함수를 분리했으므로 호출 순서도 정해져 있다.
</details>

## 8. Java 필드 숨김과 메서드 재정의
출력값은?
```java
class A { int x=1; int get(){return x;} }
class B extends A { int x=2; @Override int get(){return x;} }
public class Main { public static void main(String[] z){ A a=new B(); System.out.print(a.x+" "+a.get()); }}
```

<details><summary>정답·해설</summary>

**`1 2`**. 필드 `a.x`는 참조 변수 타입 A 기준, 오버라이딩된 `a.get()`은 실제 객체 B 기준이다.
</details>

## 11. LEFT JOIN 집계
A.id=`1,2`, B.id=`1,1`이다. 다음 SQL의 두 숫자는?
```sql
SELECT COUNT(*), COUNT(B.id)
FROM A LEFT JOIN B ON A.id=B.id;
```

<details><summary>정답·해설</summary>

**`3, 2`**. A의 1은 B 두 행과 결합해 두 행, A의 2는 B 값이 NULL인 한 행을 만든다. `COUNT(*)`는 세 행, `COUNT(B.id)`는 NULL을 제외한 두 행이다.
</details>

## 12. 상관 서브쿼리와 평균
A의 `(id,x)`는 `(1,4),(2,9),(3,15)`이고 B의 `(a_id,y)`는 `(1,2),(1,6),(2,20)`이다. 결과는?
```sql
SELECT COUNT(*) FROM A a
WHERE a.x > (SELECT AVG(b.y) FROM B b WHERE b.a_id < a.id);
```

<details><summary>정답·해설</summary>

**2**. a.id=1에서는 평균 NULL이라 제외. id=2는 id<2인 B의 평균 4보다 x=9가 크다. id=3도 B의 세 y 평균 `(2+6+20)/3≈9.33`보다 x=15가 크다.
</details>

## 13. NOT IN과 NULL
A.id=`1,2,3`, B.id=`2,NULL`이다. `SELECT COUNT(*) FROM A WHERE id NOT IN (SELECT id FROM B);`의 결과는?

<details><summary>정답·해설</summary>

**0**. 1과 3은 `2`와 다르지만 `NULL`과의 비교 결과가 UNKNOWN이라 WHERE를 통과하지 못한다. NULL을 포함할 수 있으면 `NOT EXISTS` 또는 서브쿼리의 `IS NOT NULL` 처리가 필요하다.
</details>

## 14. GROUP BY와 HAVING
부서 A의 점수는 `10,NULL,20`, 부서 B는 `30,NULL`이다. 다음 결과 행의 부서와 평균은?
```sql
SELECT 부서, AVG(점수) FROM T
GROUP BY 부서 HAVING COUNT(*) >= 3;
```

<details><summary>정답·해설</summary>

**A, 15**. `COUNT(*)`는 NULL 점수 행도 세므로 A는 3행으로 통과한다. `AVG(점수)`는 NULL을 제외해 `(10+20)/2=15`다.
</details>

## 16. SRT 동률 처리
P1 도착/실행=`0/6`, P2=`1/3`, P3=`2/2`ms다. 선점형 SRT이며 남은 시간이 같으면 현재 실행 중인 작업을 계속한다. 각 프로세스의 종료 시각을 적으시오.

<details><summary>정답·해설</summary>

**P1=11, P2=4, P3=6ms**. P1 0~1 후 P2 1~4. 시각 2에 P2 남은 2와 P3의 2가 동률이므로 P2를 계속한다. 이후 P3 4~6, P1 6~11.
</details>

## 17. 3NF와 BCNF
R(A,B,C)의 후보키가 `(A,B)`, `(A,C)`이고 함수 종속 `C→B`가 있다. C는 슈퍼키가 아니다. 3NF와 BCNF 중 무엇까지 만족하는가?

<details><summary>정답·해설</summary>

**3NF까지**. B는 후보키 `(A,B)`에 들어가는 주요 속성이므로 3NF는 가능하다. 하지만 결정자 C가 슈퍼키가 아니라 BCNF는 위반한다.
</details>

## 20. 해시와 무결성
파일을 내려받은 뒤 공개된 SHA-256 해시값과 비교하는 목적은 기밀성·무결성·가용성 중 무엇인가? 해시값만으로 발신자의 신원도 보장되는가?

<details><summary>정답·해설</summary>

**무결성 확인, 신원 보장은 아니다.** 해시가 같으면 파일 변조 여부를 점검할 수 있지만 공격자가 파일과 공개 해시값을 함께 바꿀 수 있다면 발신자 인증은 안 된다. 서명·신뢰할 수 있는 배포 경로가 별도로 필요하다.
</details>
