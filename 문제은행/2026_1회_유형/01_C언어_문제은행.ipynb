## Q1. 함수 포인터를 담은 구조체

```c
#include <stdio.h>

struct ops {
    int* (*fn)(int*);
    int base;
} mine;

int* step2(int* d) { return d + 2; }

int main() {
    int n[] = {10, 20, 30, 40};
    mine.fn = step2;
    mine.base = n[0];
    printf("%x %d", *mine.fn(n), *mine.fn(n) + mine.base);
    return 0;
}
```

**출력 결과를 쓰시오.**

````python
#n의 값을 어떻게 불러오길래 출력값에서 n을 받는건지 이해가 잘 안감
# %x는 16진수로 나타내줘야함

#10진수를 소인수 분해 했을 때 몫이 앞자리 나머지가 뒷자리인데, 나머지 같은경우 10보다 작을떄는 숫자 클떄는 10부터 a b c d e..로 진행
````

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
1e 40
```

`mine.fn` 은 `int*` 를 받아 `int*` 를 돌려주는 **함수 포인터**다. `step2` 를 담았으므로
`mine.fn(n)` 은 `n + 2`, 즉 `&n[2]` 를 반환하고 `*` 를 붙이면 **30**이다.

- `%x` 는 **16진수 소문자** 출력 → 30 = `1e`
- 두 번째는 30 + 10 = **40** (10진수)

`%x` 를 10진수로 착각해 `30 40` 이라고 쓰면 오답이다.

</details>

## Q2. 이중 포인터

```c
#include <stdio.h>

void move(int **pp) { (*pp)++; }

int main() {
    int a[5] = {1, 2, 3, 4, 5};
    int *p = a;
    move(&p);
    move(&p);
    printf("%d ", *p);
    printf("%d ", *(p + 1));
    printf("%d", *(p - 2));
    return 0;
}
```

**출력 결과를 쓰시오.**

````python
3
4
2 # 나도 왜 2적었는지를 모르겠음 4-2로 착각한듯
````

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
3 4 1
```

`move` 는 **포인터 자체를 바꾸기 위해** 포인터의 주소(`int **`)를 받는다.
`(*pp)++` 는 포인터를 한 칸 전진시킨다. 두 번 호출했으니 `p` 는 `&a[2]`.

- `*p` = a[2] = 3
- `*(p+1)` = a[3] = 4
- `*(p-2)` = a[0] = 1

`move(int *p) { p++; }` 였다면 **원본이 안 바뀌어** 답이 1 2 …이 된다.

</details>

## Q6. 구조체 배열과 포인터 증가

```c
#include <stdio.h>

struct st { char name[10]; int age; };

int main() {
    struct st s[3] = {{"Kim",20},{"Lee",25},{"Park",30}};
    struct st *p = s;
    printf("%s ", p->name);
    p++;
    printf("%d ", p->age);
    printf("%s ", (p + 1)->name);
    printf("%d", p->age + (++p)->age);
    return 0;
}
```

**출력 결과를 쓰시오.**

````python
Kim
25
Park
55
````

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
Kim 25 Park 55
```

`p++` 는 **구조체 하나 크기만큼** 전진한다.

- 처음 `p->name` = "Kim"
- `p++` 후 `p->age` = 25
- `(p+1)->name` = "Park"
- 마지막은 25 + 30 = **55**

> 참고: 마지막 줄처럼 한 식에서 `p` 를 읽고 동시에 `++p` 하는 코드는 실제로는
> 평가 순서가 정해져 있지 않다. 시험에서는 **왼쪽부터 순서대로** 계산하는 것으로 본다.

</details>
