## Q1. 얕은 복사와 += (26년1회 9번 유형)

```python
def f(a):
    m = [[x] for x in a]
    b = m[:]
    for i in range(len(b) - 1):
        b[i + 1] += b[i]
    return [len(x) for x in m]

print(f([1, 2, 3, 4, 5]))
```

**출력 결과를 쓰시오.**

````python
m = [[1],[2],[3],[4],[5]]
b = [[1],[2],[3],[4],[5]][:] >> b = [[1],[2],[3],[4],[5]]
def f(a): 
    for i in range(4):
        b[1] += b[0] >> [2,1]
        b[2] += b[1] >> [3,2,1]
        b[3] += b[2] >> [4,3,2,1]
        b[4] += b[3] >> [5,4,3,2,1]
    return [len(x) for x in m]            

#[:] 를 통해서 b값은 m과 같은 메모리주소를 바라보고 있기 때문에 결과값이 [1,2,3,4,5] 가 되는현상이 되버림
````

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
[1, 2, 3, 4, 5]
```

`m[:]` 는 **바깥 리스트만** 새로 만든다. 안쪽 리스트는 `m` 과 **같은 객체를 공유**한다.
리스트에 대한 `+=` 는 `extend` 와 같아서 **원본을 그 자리에서 변경**한다. 그래서 `b` 를 바꾸면 `m` 도 바뀐다.

- i=0 : b[1] = [2] + [1] → `[2,1]`  (m[1]도 같이 바뀜)
- i=1 : b[2] = [3] + [2,1] → `[3,2,1]`
- i=2 : b[3] → `[4,3,2,1]`
- i=3 : b[4] → `[5,4,3,2,1]`

길이는 1,2,3,4,5 가 된다.
**만약 `b[i+1] = b[i+1] + b[i]` 였다면 새 리스트가 만들어져 m은 그대로**이고 답은 `[1,1,1,1,1]` 이었다.

</details>

## Q25. 클래스 상속 + 다형성 종합

```python
class Shape:
    def __init__(self, n):
        self.n = n
    def area(self):
        return 0
    def __str__(self):
        return f"{self.n}:{self.area()}"

class Rect(Shape):
    def __init__(self, w, h):
        super().__init__("R")
        self.w, self.h = w, h
    def area(self):
        return self.w * self.h

class Square(Rect):
    def __init__(self, s):
        super().__init__(s, s)
        self.n = "S"

for s in [Shape("X"), Rect(3, 4), Square(5)]:
    print(s, end=" ")
print()
```

**출력 결과를 쓰시오.**

````python
X:0,R:12,S:25
````

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
X:0 R:12 S:25 
```

`__str__` 은 Shape에만 있지만, 그 안의 `self.area()` 는 **실제 객체의 area**를 부른다.
Square는 Rect의 생성자를 거쳐 w=h=s 가 되고, 그 뒤에 이름만 "S"로 덮어쓴다.

</details>
