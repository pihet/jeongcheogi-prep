## Q1. 얕은 복사와 += (26년1회 9번 유형)

```python
def f(a):
    m = [[x] for x in a]
    b = m[:]
    for i in range(len(b) - 1):
        b[i + 1] += b[i]
    return [len(x) for x in m]

print(f([1, 2, 3, 4, 5]))
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
[1, 2, 3, 4, 5]
```

`m[:]` 는 **바깥 리스트만** 새로 만든다. 안쪽 리스트는 `m` 과 **같은 객체를 공유**한다.
리스트에 대한 `+=` 는 `extend` 와 같아서 **원본을 그 자리에서 변경**한다. 그래서 `b` 를 바꾸면 `m` 도 바뀐다.

- i=0 : b[1] = [2] + [1] → `[2,1]`  (m[1]도 같이 바뀜)
- i=1 : b[2] = [3] + [2,1] → `[3,2,1]`
- i=2 : b[3] → `[4,3,2,1]`
- i=3 : b[4] → `[5,4,3,2,1]`

길이는 1,2,3,4,5 가 된다.
**만약 `b[i+1] = b[i+1] + b[i]` 였다면 새 리스트가 만들어져 m은 그대로**이고 답은 `[1,1,1,1,1]` 이었다.

</details>

## Q12. 상속과 super

```python
class A:
    def f(self):
        return "A" + self.g()
    def g(self):
        return "a"

class B(A):
    def g(self):
        return "b"
    def f(self):
        return super().f() + "B"

print(B().f())
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
AbB
```

`super().f()` 는 A의 f를 실행하지만, 그 안의 `self.g()` 는 **실제 객체 B의 g**를 부른다.
→ A의 f가 "A" + "b" = "Ab" 를 만들고, 뒤에 "B" 가 붙는다.

</details>

## Q17. try / except / else / finally

```python
def f(n):
    try:
        r = 10 // n
    except ZeroDivisionError:
        return "Z"
    else:
        return "E" + str(r)
    finally:
        print("F", end="")

print(f(2))
print(f(0))
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
FE5
FZ
```

`else` 는 예외가 **없을 때만** 실행된다. `finally` 는 return이 있어도 **반드시** 실행된다.
출력 순서에 주의 — finally의 F가 return 값보다 먼저 찍힌다.

</details>

## Q20. 재귀 호출 횟수

```python
cnt = 0
def fib(n):
    global cnt
    cnt += 1
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

print(fib(8), cnt)
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
21 67
```

값 자체보다 **호출 횟수**를 묻는 유형이 있다. 메모이제이션이 없으면 호출이 지수적으로 늘어난다.
호출 횟수 T(n) = T(n-1) + T(n-2) + 1 이고, 손으로 셀 때는 재귀 트리를 그린다.

</details>

## Q25. 클래스 상속 + 다형성 종합

```python
class Shape:
    def __init__(self, n):
        self.n = n
    def area(self):
        return 0
    def __str__(self):
        return f"{self.n}:{self.area()}"

class Rect(Shape):
    def __init__(self, w, h):
        super().__init__("R")
        self.w, self.h = w, h
    def area(self):
        return self.w * self.h

class Square(Rect):
    def __init__(self, s):
        super().__init__(s, s)
        self.n = "S"

for s in [Shape("X"), Rect(3, 4), Square(5)]:
    print(s, end=" ")
print()
```

**출력 결과를 쓰시오.**

## Q1. 딕셔너리 삽입 순서 (26년2회 17번 유형)

```python
locations = {
    "NYC": "NEWYORK",
    "LDN": "LONDON",
    "PAR": "PARIS",
    "TKY": "TOKYO",
}

s = ""
for key, value in locations.items():
    s += key[-1] + value[0]
print(s)
print(list(locations)[1], len(locations))
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
CNNLRPYT
LDN 4
```

**Python 3.7부터 딕셔너리는 삽입 순서를 보장한다.** 그래서 NYC → LDN → PAR → TKY 순으로 돈다.

각 항목에서 `key[-1]`(키의 마지막 글자) + `value[0]`(값의 첫 글자)를 이어붙인다.

- NYC / NEWYORK → `C` + `N`
- LDN / LONDON → `N` + `L`
- PAR / PARIS → `R` + `P`
- TKY / TOKYO → `Y` + `T`

문제에 "Python 3.7 이상"이라고 명시된 이유가 순서 보장 때문이다. 그 전 버전이면 순서가 불규칙했다.

</details>

## Q2. 슬라이싱 결합 (26년2회 6번 유형)

```python
arr = "_THIS_IS_KIM_SPEAKING"
a = arr[:4]
b = arr[6:8]
c = arr[18:]
print(a + b + c)
print(len(arr), arr[5], arr[-3:])
```

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

인덱스를 **한 칸씩 세어서** 표를 그려야 한다. 밑줄도 한 글자다.

이어붙이면 `_THIISING` 이다. `S`가 하나 더 붙은 것처럼 보이는 게 노림수다.

## Q3. 딕셔너리 순회와 정렬

```python
d = {"b": 2, "a": 1, "d": 4, "c": 3}
print(list(d.keys()))
print(sorted(d))
print(sorted(d.items(), key=lambda x: -x[1])[:2])
print(max(d, key=d.get), min(d.values()))
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
['b', 'a', 'd', 'c']
['a', 'b', 'c', 'd']
[('d', 4), ('c', 3)]
d 1
```

- `list(d.keys())` : **삽입 순서** 그대로 (정렬되지 않는다)
- `sorted(d)` : 키를 정렬한 **리스트**를 반환
- `sorted(d.items(), key=lambda x: -x[1])` : 값 기준 내림차순
- `max(d, key=d.get)` : **값이 가장 큰 키**를 반환 (값이 아니라 키다)

`sorted(d)` 가 딕셔너리가 아니라 리스트를 준다는 점, `max(d, ...)` 가 키를 준다는 점이 함정이다.

</details>

## Q4. 문자열 순회로 문자 추출

```python
words = ["Software", "Quality", "Engineer", "Test"]
head = "".join(w[0] for w in words)
tail = "".join(w[-1] for w in words)
mix = "".join(w[0] + w[-1] for w in words if len(w) > 4)
print(head)
print(tail)
print(mix, len(mix))
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
SQET
eyrt
SeQyEr 6
```

컴프리헨션 안에서 첫 글자와 마지막 글자를 뽑는다.

- `head` : 각 단어의 첫 글자
- `tail` : 각 단어의 마지막 글자
- `mix` : 길이가 4보다 **큰** 단어만 (Test는 4라서 **제외**)

`len(w) > 4` 와 `len(w) >= 4` 를 헷갈리면 Test가 들어가 답이 달라진다.

</details>
