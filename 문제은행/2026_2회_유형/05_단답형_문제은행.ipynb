## SET 1 · 문항 3 — 2026년 2회 유형 (계산형 집중)

**3.** `10.0.0.0/27` 네트워크가 다음 두 서브넷으로 나뉘어 있다.

- 첫 번째 서브넷 : `10.0.0.0` ~ `10.0.0.15`
- 두 번째 서브넷 : `10.0.0.16` ~ `10.0.0.31`

`10.0.0.1` 이 속한 서브넷의 **프리픽스 비트 길이**를 숫자로 쓰시오.

<details>
<summary><b>👉 SET 1 문항 3 정답 · 해설 보기</b></summary>

**3.** **28**

> `10.0.0.0 ~ 10.0.0.15` 는 주소가 **16개**다.
> 16 = 2⁴ 이므로 호스트 비트 4개 → 프리픽스 = 32 − 4 = **28**
>
> 문제에 적힌 `/27`에 낚이면 안 된다. /27은 **나누기 전** 원래 블록(주소 32개)이고,
> 그것을 둘로 쪼갰으니 각 서브넷은 한 비트 더 긴 **/28** 이 된다.

</details>

## SET 1 · 문항 4 — 2026년 2회 유형 (계산형 집중)

**4.** 다음 표가 만족하는 **가장 높은 정규형**을 보기에서 골라 쓰시오.

| 고객아이디 | 과정명 | 강사번호 |
|---|---|---|
| apple | 영어회화 | P001 |
| banana | 기초토익 | P002 |
| carrot | 영어회화 | P001 |
| carrot | 기초토익 | P004 |
| orange | 영어회화 | P003 |
| orange | 기초토익 | P004 |

> **보기** : 제1정규형 · 제2정규형 · 제3정규형 · 보이스코드 정규형 · 제4정규형 · 제5정규형

<details>
<summary><b>👉 SET 1 문항 4 정답 · 해설 보기</b></summary>

**4.** **제3정규형**

> 기본키는 (고객아이디, 과정명)이다.
> - **강사번호는 기본키 전체에 완전 함수 종속**된다 → 2정규형 만족
> - 비주요 속성끼리의 이행 종속이 없다 → **3정규형 만족**
> - 그런데 **강사번호 → 과정명** 이 성립한다 (P001→영어회화, P004→기초토익 …).
>   강사번호는 후보키가 아닌데 결정자 노릇을 하므로 **BCNF는 위반**한다.
>
> 따라서 만족하는 가장 높은 단계는 **제3정규형**이다.
> **BCNF = 모든 결정자가 후보키** — 이 한 줄로 판별한다.

</details>

## SET 1 · 문항 5 — 2026년 2회 유형 (계산형 집중)

**5.** 프로세스의 도착시간과 버스트 시간이 다음과 같다.
**SRT 스케줄링**을 적용할 때 **평균 대기시간**을 구하시오. (단위 ms)

| 프로세스 | 도착시간 | 버스트 시간 |
|---|---|---|
| P1 | 0 | 8 |
| P2 | 1 | 4 |
| P3 | 2 | 9 |
| P4 | 3 | 5 |

<details>
<summary><b>👉 SET 1 문항 5 정답 · 해설 보기</b></summary>

**5.** **6.5 ms**

> SRT는 **선점형**이다. 새 프로세스가 도착할 때마다 **남은 시간이 더 짧으면 뺏는다.**
>
> | 시각 | 사건 | 실행 |
> |---|---|---|
> | 0 | P1 시작 | P1 |
> | 1 | P2(4) 도착 < P1 남음(7) → **선점** | P2 |
> | 5 | P2 종료. 남은 것 P1=7, P3=9, P4=5 | P4 |
> | 10 | P4 종료 | P1 |
> | 17 | P1 종료 | P3 |
> | 26 | P3 종료 | — |
>
> **대기시간 = 완료시각 − 도착시간 − 버스트시간**
>
> | | 완료 | 도착 | 버스트 | 대기 |
> |---|---|---|---|---|
> | P1 | 17 | 0 | 8 | **9** |
> | P2 | 5 | 1 | 4 | **0** |
> | P3 | 26 | 2 | 9 | **15** |
> | P4 | 10 | 3 | 5 | **2** |
>
> (9 + 0 + 15 + 2) ÷ 4 = 26 ÷ 4 = **6.5 ms**
>
> 반드시 **간트 차트를 그리고 표로 계산**할 것. 선점 시점을 놓치면 전부 어긋난다.

</details>

## SET 1 · 문항 6 — 2026년 2회 유형 (계산형 집중)

**6.** 점수 구간별로 **대표값을 하나씩 뽑아** 테스트 케이스를 만들었다.
이때 사용된 블랙박스 테스트 기법을 보기에서 골라 쓰고, **경계값 분석과의 차이**를 한 줄로 쓰시오.

| 테스트값 | 예상 결과 |
|---|---|
| 95 | A |
| 80 | B |
| 73 | C |
| 23 | D |
| 15 | E |

> **보기** : 경계값 분석 · 동등 분할 · 결정 테이블 · 오류 예측 · 원인-결과 그래프

<details>
<summary><b>👉 SET 1 문항 6 정답 · 해설 보기</b></summary>

**6.** **동등 분할(Equivalence Partitioning) 테스트**

> 차이 — **동등 분할**은 각 구간에서 **대표값 하나**를 뽑고,
> **경계값 분석**은 구간의 **경계와 그 앞뒤 값**(예: 89, 90, 91)을 뽑는다.
>
> 위 표는 95, 80, 73, 23, 15처럼 구간 **한가운데 값**들이므로 동등 분할이다.
> 실무에서는 결함이 경계에 몰리므로 둘을 함께 쓴다.

</details>
