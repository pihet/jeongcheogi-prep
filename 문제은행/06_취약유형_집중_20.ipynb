## 1. 음수 인덱스와 역순
`s = 'TKY'`일 때 `s[-1]`, `s[::-1]`을 순서대로 적으시오.

<details><summary>정답·해설</summary>

**`Y`, `YKT`**. `[-1]`은 마지막 원소 한 개다. `[::-1]`은 슬라이스의 간격이 -1이라 전체 문자열을 역순으로 읽는다.
</details>

## 2. 딕셔너리 순회
다음 코드의 출력은?
```python
d = {'SEA': 'Seoul', 'PUS': 'Busan', 'CJU': 'Jeju'}
out = ''
for key, value in d.items():
    out += key[-1] + value[0]
print(out)
```

<details><summary>정답·해설</summary>

**`ASSBUJ`**. `SEA→AS`, `PUS→SB`, `CJU→UJ`이므로 `AS`+`SB`+`UJ`다. 각 키의 마지막 글자와 값의 첫 글자만 뽑는다.
</details>

## 3. C 포인터와 값 전달
다음 출력은?
```c
#include <stdio.h>
void f(int *p) { *p = 7; }
void g(int x) { x = 9; }
int main(void) { int n = 3; f(&n); g(n); printf("%d", n); }
```

<details><summary>정답·해설</summary>

**7**. `f`는 `n`의 주소를 받아 원본을 바꾼다. `g`는 현재 값 7의 복사본을 바꿀 뿐이다.
</details>

## 4. C 배열 포인터
`int a[] = {3, 6, 9, 12}; int *p = a;`에서 `*(p+2)`와 `p[3]`의 값을 적으시오.

<details><summary>정답·해설</summary>

**9, 12**. 포인터에 2를 더하면 정수 2개만큼 이동해 `a[2]`를 가리킨다. `p[3]`은 `*(p+3)`과 같다.
</details>

## 5. 음수까지 가는 재귀
다음 `f(4)`의 값은?
```c
int f(int n) {
    if (n <= 1) return n;
    return f(n - 1) + f(n - 3);
}
```

<details><summary>정답·해설</summary>

**1**. 종료식은 `0`을 반환하는 것이 아니라 **입력 n 그대로** 반환한다. `f(2)=f(1)+f(-1)=1-1=0`, `f(3)=f(2)+f(0)=0`, `f(4)=f(3)+f(1)=1`.
</details>

## 7. /22 네트워크 범위
`129.200.10.72/22`와 같은 네트워크에 속한 사용 가능한 호스트 주소를 하나 적고, 네트워크 주소와 브로드캐스트 주소를 구하시오.

<details><summary>정답·해설</summary>

**예: `129.200.8.249`, 네트워크 `129.200.8.0`, 브로드캐스트 `129.200.11.255`**. `/22`는 세 번째 옥텟의 블록 크기가 4라 10이 속한 시작값은 8이다. 사용 가능 범위는 `129.200.8.1~129.200.11.254`.
</details>

## 8. 두 개로 서브넷 분할
`192.168.7.64/27`을 같은 크기의 두 서브넷으로 나누면 새 프리픽스는? `192.168.7.75`가 속한 서브넷 주소도 적으시오.

<details><summary>정답·해설</summary>

**`/28`, `192.168.7.64/28`**. 2개로 나눌 때 프리픽스에 1을 더한다. `/28`은 16개 단위로 `.64~.79`, `.80~.95`가 된다.
</details>

## 9. SRT 평균 대기시간
도착/실행 시간이 P1=`0/5`, P2=`1/2`, P3=`2/1` ms일 때 선점형 SRT의 평균 대기시간을 구하시오. 동률이면 먼저 도착한 작업을 고른다.

<details><summary>정답·해설</summary>

**`4/3ms ≈ 1.33ms`**. P1 `0~1`, P2 `1~3`, P3 `3~4`, P1 `4~8`로 실행된다. P3은 시각 2에 도착해도 P2 남은 시간 1과 동률이므로 P2가 계속된다. 대기시간은 P1=`8−0−5=3`, P2=`3−1−2=0`, P3=`4−2−1=1`; 합 4다.
</details>

## 10. SQL 상관 서브쿼리
A의 x가 `10,20,30`, B의 `(id,y)`가 `(1,5),(1,15),(2,50)`이고 A의 id는 x 순으로 `1,2,3`이다. 다음 결과는?
```sql
SELECT COUNT(*) FROM A
WHERE x > (SELECT AVG(y) FROM B
           WHERE B.id IN (SELECT A2.id FROM A A2 WHERE A2.x < A.x));
```

<details><summary>정답·해설</summary>

**2**. 바깥 x=10이면 내부 행이 없어 평균 NULL이라 제외된다. x=20이면 평균 10, x=30이면 평균 `(5+15+50)/3=70/3`이므로 둘 다 조건을 만족한다.
</details>

## 11. RIGHT JOIN과 NULL
A.id=`1,1,2`, B.id=`1,3,4`다. `A RIGHT JOIN B ON A.id=B.id WHERE A.id IS NULL`의 행 수는?

<details><summary>정답·해설</summary>

**2**. B의 3과 4에 대응하는 A가 없어 남는다. B의 1은 A의 1 두 행과 결합되지만 `A.id IS NULL`이 거짓이다.
</details>

## 12. COUNT와 NULL
값이 `10, NULL, 20`인 열 `v`가 있다. `COUNT(*)`, `COUNT(v)`, `AVG(v)`를 순서대로 구하시오.

<details><summary>정답·해설</summary>

**3, 2, 15**. `COUNT(*)`는 행 수, `COUNT(v)`는 v가 NULL이 아닌 행 수다. `AVG(v)`도 NULL을 제외한 `(10+20)/2`다. 표시 자릿수는 DBMS·자료형에 따라 다르므로 임의로 반올림하지 않는다.
</details>

## 17. SQL 제약
열의 값이 `'봄','여름','가을','겨울'` 중 하나만 되도록 검사할 때 쓰는 제약 키워드는?

<details><summary>정답·해설</summary>

**CHECK**. `CHECK (계절 IN (...))`처럼 조건을 선언한다. `NOT NULL`은 NULL만 금지하므로 허용 목록을 만들지 못한다.
</details>

## 18. 정규형
함수 종속이 `(학생,과목)→강사`, `강사→과목`이고 후보키가 `(학생,과목)`, `(학생,강사)`다. 모든 속성이 원자값일 때 가장 높은 정규형은?

<details><summary>정답·해설</summary>

**3NF**. `강사→과목`의 결정자 강사가 슈퍼키가 아니므로 BCNF는 위반한다. 하지만 과목은 후보키에 포함되는 주요 속성이므로 3NF는 만족한다.
</details>

## 19. 결합도
모듈 A가 모듈 B의 내부 변수에 직접 접근해 값을 고친다. 가장 강한 이 결합도는? 공통 전역 변수를 쓰는 경우와 구분하시오.

<details><summary>정답·해설</summary>

**내용 결합도**. B의 내부 구현에 직접 접근한다. 두 모듈이 같은 전역 데이터를 공유하는 것은 공통 결합도다.
</details>

## 20. SQL 명령 분류
`CREATE`, `UPDATE`, `GRANT`, `ROLLBACK`을 각각 DDL·DML·DCL·TCL 중 하나로 분류하시오.

<details><summary>정답·해설</summary>

**CREATE=DDL, UPDATE=DML, GRANT=DCL, ROLLBACK=TCL**. 구조 정의, 행 데이터 변경, 권한 관리, 트랜잭션 제어의 차이다. 일부 교재는 TCL을 DCL에 묶으므로 시험의 분류 기준을 확인한다.
</details>
