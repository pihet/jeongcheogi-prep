## 난이도 기준 — 2026년 1·2회

이 문제은행은 **2026년 1회와 2회 실기의 출제 형식과 난이도**에 맞춰 작성했다.
두 회차의 실제 구성은 다음과 같았다.

| 유형 | 26년 1회 | 26년 2회 |
|---|---|---|
| 코딩 | 7 (C 2 · Java 2 · Python 3) | 7 (C 2 · Java 3 · Python 2) |
| SQL | 3 | **4** |
| 이론 | 10 | 9 (**계산형 4문항**) |

**두 회차의 차이 — 2회는 계산형이 크게 늘었다.**
서브네팅 프리픽스, 라우터 IP 할당, SRT 평균 대기시간, 정규형 판별까지
**손으로 표를 그려야 풀리는 문제가 4개**였다. 암기만으로는 못 넘는다.

### 코드 문제가 어려운 이유는 길이가 아니다

26년 1회 코드는 대부분 **10~25줄**로 길지 않았다. 어려운 건 함정의 **종류**였다.

| 언어 | 실제 출제된 함정 |
|---|---|
| Python | (1회) 얕은 복사 + `+=` 제자리 변경 · 음수 step 슬라이싱 / (2회) **dict 삽입 순서** · 슬라이싱 인덱스 |
| Java | (1회) 오버로딩은 **컴파일 시점** 결정 · `+` 좌결합 / (2회) **재귀의 음수 인자** · private 필드 상속 · super 호출 순서 |
| C | (1회) 구조체 안의 **함수 포인터** · `%x` 출력 / (2회) **구조체 트리 후위 순회** · 값/주소/배열 전달 비교 |

"문법을 아는가"가 아니라 **"언어가 언제 무엇을 결정하는가"**를 묻는다.
그래서 이 은행도 단일 개념 암기 문제를 전부 걷어내고 그 지점만 남겼다.

## 푸는 방법 (중요)

1. **정답 토글을 먼저 열지 마라.** `👉 정답 · 해설 보기` 는 접혀 있다.
2. 코드 문제는 **반드시 종이에 변수 표를 그려서** 추적한다. 눈으로 읽으면 틀린다.
3. 답을 적은 뒤에 토글을 연다.
4. 출력은 **공백 · 개행 · 진법 · 대소문자**까지 정확히 적는다. 부분점수가 없다.
5. 틀렸으면 **왜 틀렸는지**를 `체크리스트/오답노트_템플릿` 에 기록한다.
   (몰랐다 / 헷갈렸다 / 실수했다 / 시간 부족 — 넷 중 하나로)
6. `03_Python_문제은행` 은 코드 셀이 실행 가능하다.
   **먼저 답을 손으로 적고 나서** 실행해 채점한다. 순서를 바꾸면 공부가 안 된다.

## 정답의 신뢰도

정답은 다음 방법으로 검증했다.

- **Python 25문항** — 전부 실제로 실행해 나온 출력을 그대로 정답으로 썼다.
- **SQL 20문항** — 그중 11문항은 sqlite로 테이블을 만들어 **실제 질의를 실행**한 결과다.
- **C 25 · Java 20문항** — 컴파일러가 없으므로, 문항마다 동작을 파이썬으로 재현해
  손으로 낸 답과 **대조가 일치할 때만** 수록했다. (이 과정에서 실제로 오답 4건을 잡아냈다.)
- **계산형 단답** (서브네팅 CIDR, 호스트 수, HRN, 페이지 부재) — 코드로 계산해 확인했다.

그래도 틀린 것을 발견하면 그 자리에서 고쳐 쓰고 오답노트에 남길 것.
