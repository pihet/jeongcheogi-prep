## Q1. 함수 포인터를 담은 구조체

```c
#include <stdio.h>

struct ops {
    int* (*fn)(int*);
    int base;
} mine;

int* step2(int* d) { return d + 2; }

int main() {
    int n[] = {10, 20, 30, 40};
    mine.fn = step2;
    mine.base = n[0];
    printf("%x %d", *mine.fn(n), *mine.fn(n) + mine.base);
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
1e 40
```

`mine.fn` 은 `int*` 를 받아 `int*` 를 돌려주는 **함수 포인터**다. `step2` 를 담았으므로
`mine.fn(n)` 은 `n + 2`, 즉 `&n[2]` 를 반환하고 `*` 를 붙이면 **30**이다.

- `%x` 는 **16진수 소문자** 출력 → 30 = `1e`
- 두 번째는 30 + 10 = **40** (10진수)

`%x` 를 10진수로 착각해 `30 40` 이라고 쓰면 오답이다.

</details>

## Q2. 이중 포인터

```c
#include <stdio.h>

void move(int **pp) { (*pp)++; }

int main() {
    int a[5] = {1, 2, 3, 4, 5};
    int *p = a;
    move(&p);
    move(&p);
    printf("%d ", *p);
    printf("%d ", *(p + 1));
    printf("%d", *(p - 2));
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
3 4 1
```

`move` 는 **포인터 자체를 바꾸기 위해** 포인터의 주소(`int **`)를 받는다.
`(*pp)++` 는 포인터를 한 칸 전진시킨다. 두 번 호출했으니 `p` 는 `&a[2]`.

- `*p` = a[2] = 3
- `*(p+1)` = a[3] = 4
- `*(p-2)` = a[0] = 1

`move(int *p) { p++; }` 였다면 **원본이 안 바뀌어** 답이 1 2 …이 된다.

</details>

## Q3. 2차원 배열의 포인터 표기

```c
#include <stdio.h>

int main() {
    int a[3][3] = {{1,2,3},{4,5,6},{7,8,9}};
    printf("%d ", *(*(a + 1) + 2));
    printf("%d ", *(a[2]));
    printf("%d ", *(a[0] + 1));
    printf("%d", *(&a[0][0] + 8));
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
6 7 2 9
```

- `*(*(a+1)+2)` = `a[1][2]` = 6
- `*(a[2])` = `a[2][0]` = 7
- `*(a[0]+1)` = `a[0][1]` = 2
- `&a[0][0] + 8` : 2차원 배열은 메모리에 **연속**이므로 9번째 원소 = `a[2][2]` = 9

</details>

## Q4. 포인터 배열과 문자 접근

```c
#include <stdio.h>
#include <string.h>

int main() {
    char *p[3] = {"Data", "Base", "System"};
    printf("%s ", p[1]);
    printf("%c ", *p[2]);
    printf("%c ", *(p[0] + 2));
    printf("%d", (int)strlen(p[2]));
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
Base S t 6
```

`char *p[3]` 은 **문자열 3개를 가리키는 포인터 배열**이다.

- `p[1]` → "Base"
- `*p[2]` → p[2]의 첫 글자 'S'
- `*(p[0]+2)` → "Data"의 인덱스 2 = 't'
- `strlen("System")` = 6

`char (*p)[3]` 이라면 **열이 3개인 배열을 가리키는 포인터**로 뜻이 완전히 달라진다.

</details>

## Q5. 포인터로 문자열 순회

```c
#include <stdio.h>

int f(char *s) {
    int c = 0;
    while (*s++) c++;
    return c;
}

int g(char *s) {
    int c = 0;
    while (*s) { if (*s == 'e') c++; s++; }
    return c;
}

int main() {
    printf("%d %d", f("Engineer"), g("Engineer"));
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
8 2
```

`while (*s++)` 는 **널 문자('\0')를 만날 때까지** 도는 관용구다 → 문자열 길이 8.
`g` 는 **소문자 'e'만** 센다. "Engineer" 를 한 글자씩 보면 E, n, g, i, n, e, e, r 이므로
소문자 e는 뒤쪽 두 개뿐 → **2개**. 맨 앞의 대문자 E까지 세어 3이라고 쓰면 오답이다.

</details>

## Q6. 구조체 배열과 포인터 증가

```c
#include <stdio.h>

struct st { char name[10]; int age; };

int main() {
    struct st s[3] = {{"Kim",20},{"Lee",25},{"Park",30}};
    struct st *p = s;
    printf("%s ", p->name);
    p++;
    printf("%d ", p->age);
    printf("%s ", (p + 1)->name);
    printf("%d", p->age + (++p)->age);
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
Kim 25 Park 55
```

`p++` 는 **구조체 하나 크기만큼** 전진한다.

- 처음 `p->name` = "Kim"
- `p++` 후 `p->age` = 25
- `(p+1)->name` = "Park"
- 마지막은 25 + 30 = **55**

> 참고: 마지막 줄처럼 한 식에서 `p` 를 읽고 동시에 `++p` 하는 코드는 실제로는
> 평가 순서가 정해져 있지 않다. 시험에서는 **왼쪽부터 순서대로** 계산하는 것으로 본다.

</details>

## Q7. static 변수와 재귀 호출 횟수

```c
#include <stdio.h>

int cnt = 0;

int f(int n) {
    cnt++;
    if (n <= 0) return 0;
    return n + f(n - 1);
}

int main() {
    printf("%d %d", f(5), cnt);
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
15 6
```

`f(5)` 는 5+4+3+2+1+0 = **15**.
호출은 n = 5,4,3,2,1,0 으로 **6번** 일어난다. 종료 조건에서의 호출을 빼먹고 5라고 쓰기 쉽다.

</details>

## Q8. 비트 연산 종합

```c
#include <stdio.h>

int main() {
    int a = 13, b = 10;
    printf("%d %d %d %d %d",
           a & b, a | b, a ^ b, a << 3, (~a) & 0xFF);
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
8 15 7 104 242
```

13 = `1101`, 10 = `1010`

- `&` → `1000` = 8
- `|` → `1111` = 15
- `^` → `0111` = 7
- `13 << 3` = 13 × 2³ = **104**
- `~13` 은 -14 인데 `& 0xFF` 로 하위 8비트만 남기면 `11110010` = **242**

</details>

## Q9. 문자 산술과 아스키 코드

```c
#include <stdio.h>

int main() {
    char c = 'F';
    printf("%c %d ", c + 2, c + 2);
    printf("%c ", 'a' - 32);
    printf("%c", (c - 'A') + '0');
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
H 72 A 5
```

`'A'`=65, `'a'`=97, `'0'`=48 세 개는 반드시 외운다.

- `'F'`(70) + 2 = 72 → `%c` 로는 'H', `%d` 로는 **72**
- `'a' - 32` = 65 = 'A' (소문자→대문자)
- `'F' - 'A'` = 5, 거기에 `'0'`(48)을 더하면 53 = 문자 **'5'** (숫자→문자 변환 관용구)

</details>

## Q10. 정수 나눗셈 · 나머지 · 형변환

```c
#include <stdio.h>

int main() {
    int a = -7, b = 2;
    printf("%d %d %d ", a / b, a % b, 7 / b);
    printf("%.2f ", (float)7 / b);
    printf("%.1f", (1 + 2) / 2.0);
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
-3 -1 3 3.50 1.5
```

**C의 정수 나눗셈은 0 방향으로 버린다** → `-7 / 2` = **-3** (파이썬의 -4와 다르다).
나머지의 부호는 **왼쪽 피연산자**를 따라 `-7 % 2` = **-1**.
`(float)7 / b` 는 실수 연산이라 3.50, `(1+2)/2.0` 도 실수라 1.5.
`(1+2)/2` 였다면 정수 나눗셈으로 1이 된다.

</details>

## Q15. 구조체 포인터를 함수로 넘기기

```c
#include <stdio.h>

struct P { int x, y; };

void add5(struct P *p) { p->x += 5; (*p).y -= 5; }
void swap(struct P *p) { int t = p->x; p->x = p->y; p->y = t; }

int main() {
    struct P p = {10, 20};
    add5(&p);
    swap(&p);
    printf("%d %d", p.x, p.y);
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
15 15
```

`p->x` 와 `(*p).x` 는 **완전히 같은 표현**이다.

- add5 후 : x=15, y=15
- swap 후 : 둘 다 15라 값이 그대로

우연히 같은 값이 되는 걸 노린 문제다. 중간 과정을 표로 적지 않으면 헷갈린다.

</details>

## Q16. 함수 포인터 배열

```c
#include <stdio.h>

int add(int a, int b) { return a + b; }
int sub(int a, int b) { return a - b; }
int mul(int a, int b) { return a * b; }
int dvd(int a, int b) { return a / b; }

int main() {
    int (*f[4])(int, int) = {add, sub, mul, dvd};
    int i;
    for (i = 0; i < 4; i++) printf("%d ", f[i](12, 4));
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
16 8 48 3
```

`int (*f[4])(int, int)` 은 **함수 포인터 4개를 담은 배열**이다.
괄호를 빼고 `int *f[4](int,int)` 로 쓰면 뜻이 달라진다 — 선언 읽는 법이 그대로 출제된다.
12+4=16, 12-4=8, 12×4=48, 12/4=3.

</details>

<details>

(1,2)(1,5)(2,1)(2,4)(3,3)(4,2)(4,5)(5,1)(5,4)
→ 2 + 5 + 2 + 8 + 9 + 8 + 20 + 5 + 20 = **79**

반드시 표를 그려서 세야 한다. 암산하면 거의 틀린다.

</details>

## Q21. break 와 continue 조합

```c
#include <stdio.h>

int main() {
    int i = 0, s = 0;
    while (1) {
        i++;
        if (i > 30) break;
        if (i % 4 != 0) continue;
        if (i % 3 == 0) continue;
        s += i;
    }
    printf("%d", s);
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
76
```

4의 배수이면서 3의 배수가 **아닌** 수만 더한다.
1~30의 4의 배수는 4, 8, 12, 16, 20, 24, 28 인데 12와 24는 3의 배수라 빠진다.
→ 4 + 8 + 16 + 20 + 28 = **76**

</details>

## Q22. 배열 전달 vs 값 전달

```c
#include <stdio.h>

void f(int p[], int n) { int i; for (i = 0; i < n; i++) p[i] *= 2; }
void g(int x) { x *= 2; }

int main() {
    int a[4] = {1, 2, 3, 4};
    int b = 100, i, s = 0;
    f(a, 4);
    g(b);
    for (i = 0; i < 4; i++) s += a[i];
    printf("%d %d %d %d", a[0], a[3], s, b);
    return 0;
}
```

## Q1. 구조체 트리 후위 순회 (26년2회 13번 유형)

```c
#include <stdio.h>

typedef struct N {
    int v;
    struct N *a;
    struct N *b;
} N;

int cnt = 0;

void visit(N *n) {
    if (n == NULL) return;
    visit(n->a);
    visit(n->b);
    cnt++;
    printf("%d ", n->v);
}

int main(void) {
    N d = {7, NULL, NULL};
    N e = {9, NULL, NULL};
    N c = {53, NULL, NULL};
    N b = {12, &d, &e};
    N a = {64, &b, &c};
    visit(&a);
    printf("| %d", cnt);
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
7 9 12 53 64 | 5
```

**후위 순회(post-order)** 다 — 왼쪽 → 오른쪽 → **자기 자신** 순으로 방문한다.

트리 모양:
```
        a(64)
       /     \
    b(12)   c(53)
    /   \
  d(7)  e(9)
```

방문 순서를 손으로 따라가면 : d(7) → e(9) → b(12) → c(53) → a(64)
`cnt` 는 방문할 때마다 증가하므로 노드 수 **5**.

> 순회 종류를 구분할 것. **전위**는 자기 → 왼 → 오, **중위**는 왼 → 자기 → 오, **후위**는 왼 → 오 → 자기.
> `printf` 가 재귀 호출 **앞**에 있으면 전위, **뒤**에 있으면 후위다. 위치만 보면 바로 판별된다.

</details>

## Q2. 재귀 — 음수 인자가 들어가는 경우 (26년2회 4번 유형)

```c
#include <stdio.h>

int f(int n) {
    if (n <= 1) return n;
    return f(n - 3) + f(n - 1);
}

int main(void) {
    printf("%d %d", f(5), f(6));
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
1 1
```

함정은 `n - 3` 때문에 **인자가 음수까지 내려간다**는 것이다.
종료 조건이 `n <= 1` 이라 음수도 걸리는데, 그때 **0이 아니라 n 자신을 반환**한다.

f(5) = f(2) + f(4)
- f(2) = f(-1) + f(1) = **-1** + 1 = 0
- f(4) = f(1) + f(3) = 1 + (f(0) + f(2)) = 1 + (0 + 0) = 1
- f(5) = 0 + 1 = **1**

f(6) = f(3) + f(5) = 0 + 1 = **1**

`f(-1)` 을 0으로 착각하면 전부 틀린다. **종료 조건이 무엇을 반환하는지**를 먼저 확인할 것.

</details>

## Q3. 값 전달 · 주소 전달 · 배열 전달 (26년2회 7번 유형)

```c
#include <stdio.h>

void f1(int *p) { *p = 50; }
void f2(int p)  { p = 60; }
void f3(int p[]) { p[0] = 99; }

int main(void) {
    int i = 30;
    int arr[] = {2, 4, 6, 8, 10};
    int *q = arr;

    f1(&i);  printf("%d ", i);
    f2(i);   printf("%d ", i);
    f3(arr); printf("%d ", arr[0]);
    printf("%d %d", *q, *(q + 3));
    return 0;
}
```

- `f2(i)` : **값 복사**라 함수 안에서 60이 되어도 원본은 그대로 → **50**

## Q4. 구조체 연결 리스트 순회

```c
#include <stdio.h>

typedef struct Node {
    int v;
    struct Node *next;
} Node;

int main(void) {
    Node c = {30, NULL};
    Node b = {20, &c};
    Node a = {10, &b};
    Node *p = &a;
    int s = 0;

    while (p != NULL) {
        s += p->v;
        p = p->next;
    }
    printf("%d ", s);

    p = &a;
    printf("%d %d", p->next->v, p->next->next->v);
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
60 20 30
```

`p = p->next` 로 사슬을 따라간다. 마지막 노드의 `next` 가 `NULL` 이라 반복이 끝난다.

- 합 : 10 + 20 + 30 = **60**
- `p->next->v` = b의 값 = **20**
- `p->next->next->v` = c의 값 = **30**

선언 순서가 c → b → a 인 이유는 **뒤 노드의 주소를 먼저 알아야** 앞 노드가 가리킬 수 있기 때문이다.

</details>

## Q5. 전위 · 중위 · 후위 순회 비교

```c
#include <stdio.h>

typedef struct T { int v; struct T *l, *r; } T;

void pre(T *n)  { if (!n) return; printf("%d", n->v); pre(n->l);  pre(n->r); }
void in(T *n)   { if (!n) return; in(n->l);  printf("%d", n->v); in(n->r); }
void post(T *n) { if (!n) return; post(n->l); post(n->r); printf("%d", n->v); }

int main(void) {
    T d = {4, NULL, NULL}, e = {5, NULL, NULL}, c = {3, NULL, NULL};
    T b = {2, &d, &e};
    T a = {1, &b, &c};
    pre(&a);  printf(" ");
    in(&a);   printf(" ");
    post(&a);
    return 0;
}
```

**출력 결과를 쓰시오.**

<details>
<summary><b>👉 정답 · 해설 보기</b></summary>

```
12453 42513 45231
```

같은 트리를 세 방식으로 순회한다.
```
      1
     / \
    2   3
   / \
  4   5
```
- **전위** (자기 → 왼 → 오) : 1 2 4 5 3 → `12453`
- **중위** (왼 → 자기 → 오) : 4 2 5 1 3 → `42513`
- **후위** (왼 → 오 → 자기) : 4 5 2 3 1 → `45231`

`printf` 가 재귀 호출들 사이 **어디에 있는지**만 보면 종류를 바로 알 수 있다.

</details>
